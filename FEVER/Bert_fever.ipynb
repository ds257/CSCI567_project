```python
"""
BERT Shortcut Analysis on CLIMATE-FEVER — v2 (Improved)
=========================================================
Improvements over v1:
  Model
  -----
  • Evaluates three models: bert-base-uncased, bert-large-uncased,
    allenai/scibert_scivocab_uncased — compare shortcut behaviour across scales
    and domain adaptation.
  • Classifier head random-init warning is expected and documented.

  Training
  ---------
  • Early stopping on validation loss (patience=2) — avoids running all epochs
    when the model has already started to overfit.

  Shortcut Analysis
  ------------------
  • Lexical overlap now uses stemming (PorterStemmer) + stopword removal +
    TF-IDF weighting so "the"/"a" don't count the same as "CO2"/"methane".
  • Pearson r replaced with Spearman ρ (no linearity assumption) + Pearson r
    both reported.
  • Additional shortcut proxies:
      - Named-entity overlap (spaCy, en_core_web_sm)
      - Dependency-tree depth ratio (syntactic complexity proxy)
      - Sentence-length ratio (claim_len / evidence_len)
  • Negation detection expanded: regex covers contractions ("isn't", "won't")
    and domain negations ("absent", "fail", "false", "incorrect").

  Calibration
  ------------
  • A dedicated calibration split (10 % of data, stratified) is held out
    from both training AND evaluation so calibration metrics are unbiased.
  • Evaluation split is the remaining 10 % of data (not used for calibration).

  Class Weights
  --------------
  • Computed from the training split distribution only (not the full dataset).

Run:
  pip install torch transformers datasets scikit-learn matplotlib seaborn scipy spacy nltk
  python -m spacy download en_core_web_sm
  python -m nltk.downloader stopwords
  python bert_shortcut_analysis_v2.py
"""
# ─── Imports ──────────────────────────────────────────────────────────────────
import os, json, random, re, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
from math import log

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from torch.optim import AdamW

from datasets import load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    get_linear_schedule_with_warmup,
)
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score,
)
from sklearn.calibration import calibration_curve
from sklearn.model_selection import train_test_split
from scipy.stats import pearsonr, spearmanr

import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords

try:
    import spacy
    nlp = spacy.load("en_core_web_sm")
    SPACY_AVAILABLE = True
except Exception:
    warnings.warn("spaCy / en_core_web_sm not found. NER and depth features disabled.")
    SPACY_AVAILABLE = False

# Download NLTK data silently if missing
for pkg in ("stopwords",):
    try:
        nltk.data.find(f"corpora/{pkg}")
    except LookupError:
        nltk.download(pkg, quiet=True)

STOP_WORDS = set(stopwords.words("english"))
STEMMER    = PorterStemmer()

# ─── Config ───────────────────────────────────────────────────────────────────
SEED       = 42
MAX_LEN    = 128
BATCH_SIZE = 16
EPOCHS     = 8          # allow more epochs; early stopping will cut short
PATIENCE   = 2          # early stopping patience on val loss
LR         = 2e-5
DEVICE     = torch.device("cuda" if torch.cuda.is_available() else "cpu")

LABEL_NAMES = ["supports", "refutes", "not_enough_info"]

# FIX: evaluate multiple models to see if shortcut behaviour is model-specific
MODELS = {
    "bert-base":   "bert-base-uncased",
    "bert-large":  "bert-large-uncased",
    "scibert":     "allenai/scibert_scivocab_uncased",
}

OUTPUT_DIR = "./outputs_v2"
os.makedirs(OUTPUT_DIR, exist_ok=True)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


# ══════════════════════════════════════════════════════════════════════════════
# 1. Data Loading & Splitting
# ══════════════════════════════════════════════════════════════════════════════
def load_climate_fever():
    print("Loading CLIMATE-FEVER from HuggingFace...")
    raw = load_dataset("tdiggelm/climate_fever", split="test")
    print(f"  Raw claims: {len(raw)}")
    rows = []
    for item in raw:
        claim = item["claim"]
        for ev in item["evidences"]:
            label = ev["evidence_label"]
            if label == 3:
                continue
            rows.append({
                "claim":    claim,
                "evidence": ev["evidence"],
                "label":    label,
            })
    df = pd.DataFrame(rows)
    print(f"  Expanded pairs: {len(df)}")
    counts = Counter(df["label"])
    for i, name in enumerate(LABEL_NAMES):
        n = counts[i]
        print(f"    {name:20s}: {n:5d} ({n/len(df)*100:.1f}%)")
    return df, counts


def split_data(df):
    """
    Stratified three-way split:
      train  80 %  — fine-tuning
      cal    10 %  — temperature calibration ONLY (never used for eval)
      val    10 %  — final evaluation ONLY (never seen by calibration)
    """
    train_df, temp_df = train_test_split(
        df, test_size=0.20, stratify=df["label"], random_state=SEED
    )
    cal_df, val_df = train_test_split(
        temp_df, test_size=0.50, stratify=temp_df["label"], random_state=SEED
    )
    for name, split in [("Train", train_df), ("Cal", cal_df), ("Val", val_df)]:
        print(f"  {name}: {len(split)}")
    return (
        train_df.reset_index(drop=True),
        cal_df.reset_index(drop=True),
        val_df.reset_index(drop=True),
    )


# ══════════════════════════════════════════════════════════════════════════════
# 2. PyTorch Dataset
# ══════════════════════════════════════════════════════════════════════════════
class ClimateFEVERDataset(Dataset):
    def __init__(self, df, tokenizer, max_len=MAX_LEN):
        self.df        = df
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row   = self.df.iloc[idx]
        claim = str(row["claim"])
        evid  = str(row["evidence"])
        label = int(row["label"])
        enc = self.tokenizer(
            claim, evid,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_tensors="pt",
        )
        return {
            "input_ids":      enc["input_ids"].squeeze(),
            "attention_mask": enc["attention_mask"].squeeze(),
            "token_type_ids": enc.get("token_type_ids",
                              torch.zeros(self.max_len, dtype=torch.long)).squeeze(),
            "label":   torch.tensor(label, dtype=torch.long),
            "claim":   claim,
            "evidence": evid,
        }


# ══════════════════════════════════════════════════════════════════════════════
# 3. Shortcut Feature Functions (improved)
# ══════════════════════════════════════════════════════════════════════════════

def _stem_filter(text: str) -> list[str]:
    """Lowercase → remove stopwords → stem."""
    tokens = re.findall(r"\b\w+\b", text.lower())
    return [STEMMER.stem(t) for t in tokens if t not in STOP_WORDS]


def _tfidf_weights(corpus_tokens: list[list[str]]) -> dict[str, float]:
    """
    Compute IDF over a list of token lists.
    Returns {stem: idf_weight}.
    """
    N   = len(corpus_tokens)
    df  = Counter()
    for tokens in corpus_tokens:
        df.update(set(tokens))
    return {term: log((N + 1) / (cnt + 1)) + 1.0 for term, cnt in df.items()}


def tfidf_overlap(claim: str, evidence: str, idf: dict[str, float]) -> float:
    """
    TF-IDF weighted lexical overlap:
      Σ idf(w) for w in stem(evid) ∩ stem(claim)
      ─────────────────────────────────────────────
      Σ idf(w) for w in stem(evid)
    Falls back to 0 for empty evidence.
    """
    c_stems = set(_stem_filter(claim))
    e_stems = _stem_filter(evidence)
    if not e_stems:
        return 0.0
    denom = sum(idf.get(w, 1.0) for w in e_stems)
    numer = sum(idf.get(w, 1.0) for w in e_stems if w in c_stems)
    return numer / denom if denom > 0 else 0.0


def has_negation(text: str) -> int:
    """
    Expanded negation detection:
      - standard negation words
      - contracted forms (isn't, won't, can't …)
      - domain negations (absent, fail, false, incorrect, …)
    """
    text_l = text.lower()
    # contracted forms
    if re.search(r"\b\w+n't\b", text_l):
        return 1
    neg_words = {
        "not", "no", "never", "neither", "nor",
        "nobody", "nothing", "none", "without",
        "lack", "lacking", "absent", "fail", "fails",
        "failed", "false", "incorrect", "untrue", "deny",
        "denies", "denied", "dispute", "disputed",
    }
    return int(bool(set(re.findall(r"\b\w+\b", text_l)) & neg_words))


def ner_overlap(claim: str, evidence: str) -> float:
    """
    Named-entity overlap: fraction of claim NE spans present in evidence.
    Requires spaCy.  Returns NaN if unavailable.
    """
    if not SPACY_AVAILABLE:
        return float("nan")
    c_ents = {e.text.lower() for e in nlp(claim).ents}
    e_text = evidence.lower()
    if not c_ents:
        return 0.0
    return sum(1 for ent in c_ents if ent in e_text) / len(c_ents)


def dep_depth_ratio(claim: str, evidence: str) -> float:
    """
    Ratio of parse-tree depths: claim_depth / evidence_depth.
    Higher ratio → claim is syntactically more complex relative to evidence.
    Requires spaCy.  Returns NaN if unavailable.
    """
    if not SPACY_AVAILABLE:
        return float("nan")

    def max_depth(doc):
        depths = {}
        for token in doc:
            d, cur = 0, token
            while cur.head != cur:
                cur = cur.head
                d  += 1
                if d > 50:  # guard against malformed trees
                    break
            depths[token.i] = d
        return max(depths.values(), default=0)

    cd = max_depth(nlp(claim))
    ed = max_depth(nlp(evidence))
    return cd / ed if ed > 0 else float("nan")


def compute_shortcut_features(df: pd.DataFrame, idf: dict) -> pd.DataFrame:
    out = df.copy()
    out["overlap"]          = out.apply(
        lambda r: tfidf_overlap(r["claim"], r["evidence"], idf), axis=1)
    out["negation"]         = out["evidence"].apply(has_negation)
    out["claim_len"]        = out["claim"].apply(lambda x: len(x.split()))
    out["evid_len"]         = out["evidence"].apply(lambda x: len(x.split()))
    out["len_ratio"]        = out["claim_len"] / out["evid_len"].replace(0, np.nan)
    out["ner_overlap"]      = out.apply(
        lambda r: ner_overlap(r["claim"], r["evidence"]), axis=1)
    out["dep_depth_ratio"]  = out.apply(
        lambda r: dep_depth_ratio(r["claim"], r["evidence"]), axis=1)
    out["label_name"]       = out["label"].map({i: n for i, n in enumerate(LABEL_NAMES)})
    return out


def build_idf(df: pd.DataFrame) -> dict:
    """Build IDF from the training set only."""
    corpus = [_stem_filter(r["claim"] + " " + r["evidence"])
              for _, r in df.iterrows()]
    return _tfidf_weights(corpus)


# ══════════════════════════════════════════════════════════════════════════════
# 4. Training & Evaluation (with Early Stopping)
# ══════════════════════════════════════════════════════════════════════════════
def train_epoch(model, loader, optimizer, scheduler, criterion, device):
    model.train()
    total_loss = correct = total = 0
    for batch in loader:
        ids   = batch["input_ids"].to(device)
        mask  = batch["attention_mask"].to(device)
        ttype = batch["token_type_ids"].to(device)
        labs  = batch["label"].to(device)
        optimizer.zero_grad()
        logits = model(input_ids=ids, attention_mask=mask,
                       token_type_ids=ttype).logits
        loss   = criterion(logits, labs)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        correct    += (logits.argmax(-1) == labs).sum().item()
        total      += labs.size(0)
    return total_loss / len(loader), correct / total


def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss = correct = total = 0
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for batch in loader:
            ids   = batch["input_ids"].to(device)
            mask  = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)
            labs  = batch["label"].to(device)
            logits = model(input_ids=ids, attention_mask=mask,
                           token_type_ids=ttype).logits
            probs  = torch.softmax(logits, dim=-1)
            loss   = criterion(logits, labs)
            total_loss += loss.item()
            preds       = logits.argmax(-1)
            correct    += (preds == labs).sum().item()
            total      += labs.size(0)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labs.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())
    return (
        total_loss / len(loader),
        correct / total,
        np.array(all_preds),
        np.array(all_labels),
        np.array(all_probs),
    )


def run_training(model_name, train_ds, val_ds, class_weights=None, tag="baseline"):
    """
    Fine-tune a transformer on CLIMATE-FEVER.
    FIX: early stopping on validation loss (patience=PATIENCE).
    FIX: class weights computed from training split only (passed in).
    """
    print(f"\n{'='*60}")
    print(f"  Training: {tag}  [{model_name}]")
    print(f"{'='*60}")

    model = AutoModelForSequenceClassification.from_pretrained(
        model_name, num_labels=3, ignore_mismatched_sizes=True
    ).to(DEVICE)

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=0)
    val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    criterion = (nn.CrossEntropyLoss(weight=class_weights.to(DEVICE))
                 if class_weights is not None else nn.CrossEntropyLoss())
    if class_weights is not None:
        print(f"  Class weights (train split): {[f'{w:.3f}' for w in class_weights.tolist()]}")

    optimizer   = AdamW(model.parameters(), lr=LR, weight_decay=0.01)
    total_steps = len(train_loader) * EPOCHS
    scheduler   = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=int(0.1 * total_steps),
        num_training_steps=total_steps,
    )

    # Early stopping state
    best_val_loss   = float("inf")
    patience_count  = 0
    best_state      = None
    history         = []

    for epoch in range(1, EPOCHS + 1):
        tr_loss, tr_acc = train_epoch(
            model, train_loader, optimizer, scheduler, criterion, DEVICE
        )
        vl_loss, vl_acc, vl_preds, vl_labels, vl_probs = evaluate(
            model, val_loader, criterion, DEVICE
        )
        history.append(dict(epoch=epoch, tr_loss=tr_loss, tr_acc=tr_acc,
                            vl_loss=vl_loss, vl_acc=vl_acc))
        print(
            f"  Epoch {epoch}/{EPOCHS} | "
            f"Train Loss {tr_loss:.4f} Acc {tr_acc:.4f} | "
            f"Val Loss {vl_loss:.4f} Acc {vl_acc:.4f}"
        )

        # FIX: early stopping check
        if vl_loss < best_val_loss - 1e-4:
            best_val_loss  = vl_loss
            patience_count = 0
            best_state     = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            print(f"    ↓ New best val loss ({best_val_loss:.4f}) — checkpoint saved.")
        else:
            patience_count += 1
            print(f"    No improvement ({patience_count}/{PATIENCE})")
            if patience_count >= PATIENCE:
                print(f"  Early stopping at epoch {epoch}.")
                break

    # Restore best checkpoint
    if best_state is not None:
        model.load_state_dict({k: v.to(DEVICE) for k, v in best_state.items()})
        print("  Best checkpoint restored.")

    # Final eval on best model
    _, _, vl_preds, vl_labels, vl_probs = evaluate(
        model, val_loader, criterion, DEVICE
    )

    save_path = os.path.join(OUTPUT_DIR, f"bert_{tag}")
    model.save_pretrained(save_path)
    print(f"  Model saved → {save_path}")
    return model, vl_preds, vl_labels, vl_probs, history


# ══════════════════════════════════════════════════════════════════════════════
# 5. Temperature Calibration (on separate cal split)
# ══════════════════════════════════════════════════════════════════════════════
class TemperatureScaler(nn.Module):
    """
    FIX: fit on a dedicated calibration split; never on the evaluation split.
    """
    def __init__(self):
        super().__init__()
        self.temperature = nn.Parameter(torch.ones(1))

    def forward(self, logits):
        return logits / self.temperature

    def fit(self, model, cal_loader, device, lr=0.01, max_iter=50):
        self.cpu()
        model.eval()
        logits_list, labels_list = [], []
        with torch.no_grad():
            for batch in cal_loader:
                ids   = batch["input_ids"].to(device)
                mask  = batch["attention_mask"].to(device)
                ttype = batch["token_type_ids"].to(device)
                labs  = batch["label"].to(device)
                out   = model(input_ids=ids, attention_mask=mask,
                              token_type_ids=ttype)
                logits_list.append(out.logits.cpu())
                labels_list.append(labs.cpu())
        logits_all = torch.cat(logits_list)
        labels_all = torch.cat(labels_list)
        criterion  = nn.CrossEntropyLoss()
        opt        = torch.optim.LBFGS([self.temperature], lr=lr, max_iter=max_iter)

        def step():
            opt.zero_grad()
            loss = criterion(self.forward(logits_all), labels_all)
            loss.backward()
            return loss

        opt.step(step)
        T = self.temperature.item()
        print(f"  Learned temperature T = {T:.4f}")

        # Produce calibrated predictions for the EVAL (val) split — not cal
        return T

    def calibrate(self, model, eval_loader, device):
        """Apply learned T to the separate evaluation split."""
        self.cpu()
        model.eval()
        logits_list, labels_list = [], []
        with torch.no_grad():
            for batch in eval_loader:
                ids   = batch["input_ids"].to(device)
                mask  = batch["attention_mask"].to(device)
                ttype = batch["token_type_ids"].to(device)
                labs  = batch["label"].to(device)
                out   = model(input_ids=ids, attention_mask=mask,
                              token_type_ids=ttype)
                logits_list.append(out.logits.cpu())
                labels_list.append(labs.cpu())
        logits_all = torch.cat(logits_list)
        labels_all = torch.cat(labels_list)
        cal_probs  = torch.softmax(self.forward(logits_all).detach(), dim=-1).numpy()
        cal_preds  = cal_probs.argmax(axis=-1)
        return cal_preds, labels_all.numpy(), cal_probs


# ══════════════════════════════════════════════════════════════════════════════
# 6. Shortcut Analysis (improved)
# ══════════════════════════════════════════════════════════════════════════════
def shortcut_analysis(model, val_loader, val_df, idf, device, tag="baseline"):
    """
    FIX: Spearman ρ (no linearity assumption) in addition to Pearson r.
    FIX: TF-IDF weighted overlap.
    FIX: NER overlap + dep depth ratio + len ratio as additional proxies.
    """
    model.eval()
    rows = []
    with torch.no_grad():
        for batch in val_loader:
            ids    = batch["input_ids"].to(device)
            mask   = batch["attention_mask"].to(device)
            ttype  = batch["token_type_ids"].to(device)
            labs   = batch["label"].to(device)
            claims = batch["claim"]
            evids  = batch["evidence"]
            logits = model(input_ids=ids, attention_mask=mask,
                           token_type_ids=ttype).logits
            probs  = torch.softmax(logits, dim=-1).cpu().numpy()
            preds  = probs.argmax(axis=-1)
            labs_np = labs.cpu().numpy()
            for i in range(len(labs_np)):
                rows.append({
                    "label":      labs_np[i],
                    "pred":       preds[i],
                    "confidence": float(probs[i].max()),
                    "overlap":    tfidf_overlap(claims[i], evids[i], idf),
                    "ner_ov":     ner_overlap(claims[i], evids[i]),
                    "len_ratio":  (len(claims[i].split()) /
                                   max(len(evids[i].split()), 1)),
                    "negation":   has_negation(evids[i]),
                    "correct":    int(preds[i] == labs_np[i]),
                    "tag":        tag,
                })
    df = pd.DataFrame(rows)
    df["label_name"] = df["label"].map({i: n for i, n in enumerate(LABEL_NAMES)})

    # FIX: both Pearson r and Spearman ρ
    r,  pr = pearsonr(df["overlap"], df["confidence"])
    rho, ps = spearmanr(df["overlap"], df["confidence"])
    print(f"\n[{tag}] TF-IDF Overlap–Confidence:")
    print(f"  Pearson  r = {r:.4f}  (p={pr:.2e})")
    print(f"  Spearman ρ = {rho:.4f}  (p={ps:.2e})")
    if abs(rho) > 0.1:
        print("  ⚠  Non-trivial monotonic relationship — shortcut signal!")

    # NER overlap correlation (if available)
    valid_ner = df["ner_ov"].dropna()
    if len(valid_ner) > 10:
        rho_ner, _ = spearmanr(valid_ner, df.loc[valid_ner.index, "confidence"])
        print(f"  NER Overlap Spearman ρ = {rho_ner:.4f}")

    # Length ratio correlation
    rho_len, _ = spearmanr(df["len_ratio"].fillna(1), df["confidence"])
    print(f"  Length Ratio Spearman ρ = {rho_len:.4f}")

    print(f"\n[{tag}] Accuracy by TF-IDF overlap quartile:")
    df["overlap_quartile"] = pd.qcut(
        df["overlap"], q=4, labels=["Q1 (low)", "Q2", "Q3", "Q4 (high)"]
    )
    print(df.groupby("overlap_quartile")["correct"].mean().to_string())

    print(f"\n[{tag}] Accuracy by class:")
    print(df.groupby("label_name")["correct"].mean().to_string())

    return df, r, rho


# ══════════════════════════════════════════════════════════════════════════════
# 7. Visualisation — extended figure
# ══════════════════════════════════════════════════════════════════════════════
def plot_all(
    raw_df,
    base_df, base_preds, base_labels, base_probs,
    wt_df,   wt_preds,   wt_labels,   wt_probs,
    cal_preds, cal_labels, cal_probs,
    base_r, base_rho, wt_r, wt_rho,
    model_results,   # dict: model_name → {accuracy, f1}
    train_histories, # dict: tag → list of epoch dicts
):
    COLORS = ["steelblue", "tomato", "orange"]
    fig, axes = plt.subplots(3, 3, figsize=(18, 16))
    fig.suptitle(
        "BERT Shortcut Analysis on CLIMATE-FEVER — v2 (Improved)\n"
        "Fixes: TF-IDF overlap · Spearman ρ · NER/len features · "
        "Early stopping · Separate cal split · Train-split class weights",
        fontsize=12, fontweight="bold",
    )

    # ── 1: Label distribution ─────────────────────────────────────────────────
    ax = axes[0, 0]
    counts = [Counter(raw_df["label"])[i] for i in range(3)]
    bars   = ax.bar(LABEL_NAMES, counts, color=COLORS)
    for b, c in zip(bars, counts):
        ax.text(b.get_x() + b.get_width()/2, b.get_height() + 10,
                str(c), ha="center", fontsize=9)
    ax.set_title("Label Distribution (full dataset)")
    ax.set_ylabel("Count")
    ax.tick_params(axis="x", rotation=15)

    # ── 2: TF-IDF overlap by label ─────────────────────────────────────────────
    ax = axes[0, 1]
    for name, color in zip(LABEL_NAMES, COLORS):
        sub = raw_df[raw_df["label_name"] == name]["overlap"]
        ax.hist(sub, bins=30, alpha=0.6, label=name, color=color, density=True)
    ax.set_title("TF-IDF Weighted Overlap by Label\n"
                 "(stopwords & stems removed — more meaningful than raw overlap)")
    ax.set_xlabel("TF-IDF Overlap Score")
    ax.set_ylabel("Density")
    ax.legend(fontsize=8)

    # ── 3: Confidence vs overlap (baseline) — Spearman ρ in title ─────────────
    ax = axes[0, 2]
    for name, color in zip(LABEL_NAMES, COLORS):
        sub = base_df[base_df["label_name"] == name]
        ax.scatter(sub["overlap"], sub["confidence"],
                   alpha=0.25, s=8, label=name, color=color)
    ax.set_title(f"Baseline: Confidence vs. TF-IDF Overlap\n"
                 f"Pearson r={base_r:.3f}  Spearman ρ={base_rho:.3f}")
    ax.set_xlabel("TF-IDF Overlap")
    ax.set_ylabel("Confidence")
    ax.legend(markerscale=3, fontsize=8)

    # ── 4: Accuracy by overlap quartile: baseline vs weighted ─────────────────
    ax = axes[1, 0]
    bq = base_df.groupby("overlap_quartile")["correct"].mean()
    wq = wt_df.groupby("overlap_quartile")["correct"].mean()
    x  = np.arange(4)
    ax.bar(x - 0.2, bq.values, 0.4, label="Baseline",       color="steelblue")
    ax.bar(x + 0.2, wq.values, 0.4, label="Class-Weighted",  color="coral")
    ax.set_xticks(x)
    ax.set_xticklabels(["Q1\n(low)", "Q2", "Q3", "Q4\n(high)"])
    ax.set_title("Accuracy by TF-IDF Overlap Quartile")
    ax.set_ylabel("Accuracy")
    ax.legend()
    ax.set_ylim(0, 1)

    # ── 5: Confusion matrix — baseline ────────────────────────────────────────
    ax = axes[1, 1]
    cm = confusion_matrix(base_labels, base_preds)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=ax,
                xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES)
    ax.set_title("Confusion Matrix: Baseline")
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    ax.tick_params(axis="x", rotation=20)

    # ── 6: Per-class F1 across methods ────────────────────────────────────────
    ax = axes[1, 2]
    base_f1 = f1_score(base_labels, base_preds, average=None, labels=[0,1,2])
    wt_f1   = f1_score(wt_labels,   wt_preds,   average=None, labels=[0,1,2])
    cal_f1  = f1_score(cal_labels,  cal_preds,  average=None, labels=[0,1,2])
    x = np.arange(3)
    ax.bar(x - 0.27, base_f1, 0.27, label="Baseline",       color="steelblue")
    ax.bar(x,        wt_f1,   0.27, label="Class-Weighted",  color="coral")
    ax.bar(x + 0.27, cal_f1,  0.27, label="Calibrated (sep. split)", color="mediumseagreen")
    ax.set_xticks(x)
    ax.set_xticklabels(LABEL_NAMES, rotation=10)
    ax.set_title("Per-Class F1: All Three Methods")
    ax.set_ylabel("F1 Score")
    ax.legend(fontsize=8)
    ax.set_ylim(0, 1)

    # ── 7: Model comparison bar (if multiple models trained) ──────────────────
    ax = axes[2, 0]
    if model_results:
        mod_names  = list(model_results.keys())
        mod_accs   = [model_results[m]["accuracy"] for m in mod_names]
        mod_f1_nei = [model_results[m]["f1_nei"]   for m in mod_names]
        x = np.arange(len(mod_names))
        ax.bar(x - 0.2, mod_accs,   0.4, label="Accuracy",  color="steelblue")
        ax.bar(x + 0.2, mod_f1_nei, 0.4, label="NEI F1",    color="tomato")
        ax.set_xticks(x)
        ax.set_xticklabels(mod_names, rotation=10, fontsize=9)
        ax.set_title("Model Comparison: bert-base vs bert-large vs SciBERT")
        ax.set_ylabel("Score")
        ax.legend()
        ax.set_ylim(0, 1)
    else:
        ax.text(0.5, 0.5, "Model comparison\n(run all three models)",
                ha="center", va="center", transform=ax.transAxes, fontsize=10)
        ax.set_title("Model Comparison (bert-base / bert-large / SciBERT)")

    # ── 8: Calibration curve — baseline ───────────────────────────────────────
    ax = axes[2, 1]
    for cls_id, (name, color) in enumerate(zip(LABEL_NAMES, COLORS)):
        bl = (base_labels == cls_id).astype(int)
        try:
            fp, mp = calibration_curve(bl, base_probs[:, cls_id], n_bins=10)
            ax.plot(mp, fp, marker="o", label=name, color=color)
        except ValueError:
            pass
    ax.plot([0, 1], [0, 1], "k--", label="Perfect")
    ax.set_title("Calibration: Baseline\n(above diagonal = overconfident)")
    ax.set_xlabel("Mean Predicted Probability")
    ax.set_ylabel("Fraction of Positives")
    ax.legend(fontsize=8)

    # ── 9: Calibration curve — after temp scaling (on separate eval split) ────
    ax = axes[2, 2]
    for cls_id, (name, color) in enumerate(zip(LABEL_NAMES, COLORS)):
        bl = (cal_labels == cls_id).astype(int)
        try:
            fp, mp = calibration_curve(bl, cal_probs[:, cls_id], n_bins=10)
            ax.plot(mp, fp, marker="o", label=name, color=color)
        except ValueError:
            pass
    ax.plot([0, 1], [0, 1], "k--", label="Perfect")
    ax.set_title("Calibration: After Temp Scaling\n"
                 "(evaluated on SEPARATE held-out split)")
    ax.set_xlabel("Mean Predicted Probability")
    ax.set_ylabel("Fraction of Positives")
    ax.legend(fontsize=8)

    plt.tight_layout()
    out = os.path.join(OUTPUT_DIR, "shortcut_analysis_v2.png")
    plt.savefig(out, dpi=150, bbox_inches="tight")
    print(f"\nFigure saved → {out}")
    plt.show()


# ══════════════════════════════════════════════════════════════════════════════
# 8. Main
# ══════════════════════════════════════════════════════════════════════════════
def main():
    print(f"Device: {DEVICE}\n")

    # ── Data ──────────────────────────────────────────────────────────────────
    full_df, _       = load_climate_fever()
    train_df, cal_df, val_df = split_data(full_df)

    # FIX: build IDF from training split only
    print("\nBuilding TF-IDF weights from training split...")
    idf = build_idf(train_df)

    # FIX: class weights from training split only
    train_counts = Counter(train_df["label"].tolist())
    total_train  = sum(train_counts.values())
    class_weights = torch.tensor(
        [total_train / (3 * train_counts[i]) for i in range(3)], dtype=torch.float
    )
    print(f"Class weights (train split): {[f'{w:.3f}' for w in class_weights.tolist()]}")

    # Compute shortcut features on val split
    print("\nComputing shortcut features on validation set...")
    raw_df = compute_shortcut_features(val_df, idf)
    print("  Mean TF-IDF overlap by label:")
    print(raw_df.groupby("label_name")["overlap"].mean().to_string())

    # Tokenizer shared for bert-base runs; updated per model in multi-model loop
    primary_model_key = "bert-base"
    primary_model_name = MODELS[primary_model_key]
    tokenizer = AutoTokenizer.from_pretrained(primary_model_name)

    train_ds  = ClimateFEVERDataset(train_df, tokenizer)
    cal_ds    = ClimateFEVERDataset(cal_df,   tokenizer)
    val_ds    = ClimateFEVERDataset(val_df,   tokenizer)

    cal_loader = DataLoader(cal_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    # ── Primary runs (bert-base: baseline + class-weighted) ───────────────────
    base_model, base_preds, base_labels, base_probs, base_hist = run_training(
        primary_model_name, train_ds, val_ds,
        class_weights=None, tag="baseline"
    )
    wt_model, wt_preds, wt_labels, wt_probs, wt_hist = run_training(
        primary_model_name, train_ds, val_ds,
        class_weights=class_weights, tag="class_weighted"
    )

    # FIX: temperature calibration on dedicated cal split; eval on val split
    print("\nFitting temperature scaling on calibration split...")
    scaler = TemperatureScaler()
    T = scaler.fit(base_model, cal_loader, DEVICE)
    cal_preds, cal_labels, cal_probs = scaler.calibrate(base_model, val_loader, DEVICE)

    # ── Shortcut analysis ──────────────────────────────────────────────────────
    print("\nShortcut analysis — baseline:")
    base_df, base_r, base_rho = shortcut_analysis(
        base_model, val_loader, val_df, idf, DEVICE, "baseline"
    )
    print("\nShortcut analysis — class-weighted:")
    wt_df, wt_r, wt_rho = shortcut_analysis(
        wt_model, val_loader, val_df, idf, DEVICE, "class_weighted"
    )

    # ── Multi-model comparison (bert-large, SciBERT) ─────────────────────────
    model_results = {}
    # Always record bert-base result
    model_results["bert-base"] = {
        "accuracy": accuracy_score(base_labels, base_preds),
        "f1_nei":   f1_score(base_labels, base_preds, average=None,
                             labels=[0,1,2])[2],
    }
    for key, mname in MODELS.items():
        if key == "bert-base":
            continue
        try:
            print(f"\n[Multi-model] Training {key} ({mname})...")
            tok_m  = AutoTokenizer.from_pretrained(mname)
            tds_m  = ClimateFEVERDataset(train_df, tok_m)
            vds_m  = ClimateFEVERDataset(val_df,   tok_m)
            m_model, m_preds, m_labels, _, _ = run_training(
                mname, tds_m, vds_m,
                class_weights=class_weights, tag=key
            )
            model_results[key] = {
                "accuracy": accuracy_score(m_labels, m_preds),
                "f1_nei":   f1_score(m_labels, m_preds, average=None,
                                     labels=[0,1,2])[2],
            }
        except Exception as e:
            print(f"  Skipping {key}: {e}")

    # ── Classification reports ────────────────────────────────────────────────
    for tag, labs, preds in [
        ("Baseline",                        base_labels, base_preds),
        ("Class-Weighted",                  wt_labels,   wt_preds),
        ("Baseline + Temp Calibration",     cal_labels,  cal_preds),
    ]:
        print(f"\n{'─'*55}")
        print(f"  {tag}")
        print(f"  Overall Accuracy: {accuracy_score(labs, preds):.4f}")
        print(classification_report(labs, preds, target_names=LABEL_NAMES))

    # ── Save JSON summary ──────────────────────────────────────────────────────
    results = {
        "dataset":      "CLIMATE-FEVER",
        "total_pairs":  len(full_df),
        "train_n":      len(train_df),
        "cal_n":        len(cal_df),
        "val_n":        len(val_df),
        "baseline": {
            "accuracy":         accuracy_score(base_labels, base_preds),
            "f1_per_class":     f1_score(base_labels, base_preds,
                                         average=None, labels=[0,1,2]).tolist(),
            "overlap_pearsonr": base_r,
            "overlap_spearman": base_rho,
        },
        "class_weighted": {
            "accuracy":         accuracy_score(wt_labels, wt_preds),
            "f1_per_class":     f1_score(wt_labels, wt_preds,
                                         average=None, labels=[0,1,2]).tolist(),
            "overlap_pearsonr": wt_r,
            "overlap_spearman": wt_rho,
        },
        "calibration": {
            "learned_temperature":     T,
            "cal_split_size":          len(cal_df),
            "note":  "fit on cal split; evaluated on separate val split",
            "accuracy":    accuracy_score(cal_labels, cal_preds),
            "f1_per_class": f1_score(cal_labels, cal_preds,
                                     average=None, labels=[0,1,2]).tolist(),
        },
        "model_comparison": model_results,
    }
    json_path = os.path.join(OUTPUT_DIR, "results_summary_v2.json")
    with open(json_path, "w") as f:
        json.dump(results, f, indent=2)
    print(f"\nResults JSON → {json_path}")

    # ── Plots ─────────────────────────────────────────────────────────────────
    plot_all(
        raw_df,
        base_df, base_preds, base_labels, base_probs,
        wt_df,   wt_preds,   wt_labels,   wt_probs,
        cal_preds, cal_labels, cal_probs,
        base_r, base_rho, wt_r, wt_rho,
        model_results,
        {"baseline": base_hist, "class_weighted": wt_hist},
    )
    print("\n✓ All done. Outputs saved to ./outputs_v2/")


if __name__ == "__main__":
    main()
```
