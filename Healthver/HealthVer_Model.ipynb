```python
"""
BERT Shortcut Analysis on HealthVer
====================================
Hypothesis: BERT may rely on lexical overlap shortcuts rather than genuine
semantic understanding when verifying health-related claims.

Dataset: HealthVer
  - claim / evidence / label format
  - Labels: SUPPORTS (0) / REFUTES (1) / NEUTRAL (2)
  - We train on the provided train split and evaluate on dev/test

Structure:
  1. Load HealthVer CSVs into (claim, evidence, label) triples
  2. Baseline BERT fine-tuning (standard cross-entropy)
  3. Shortcut analysis (lexical overlap vs. model confidence, per label)
  4. Mitigation 1: Class-weighted training
  5. Mitigation 2: Temperature calibration (post-hoc)
  6. Before vs. After comparison with a 9-panel figure
"""

# ─── Imports ──────────────────────────────────────────────────────────────────
import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from torch.optim import AdamW

from transformers import (
    BertTokenizer,
    BertForSequenceClassification,
    get_linear_schedule_with_warmup,
)
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score,
)
from sklearn.calibration import calibration_curve
from sklearn.model_selection import train_test_split
from scipy.stats import pearsonr

# ─── Config ───────────────────────────────────────────────────────────────────
SEED       = 42
MAX_LEN    = 128
BATCH_SIZE = 16
EPOCHS     = 4
LR         = 2e-5
DEVICE     = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# HealthVer label mapping
# 0 = supports
# 1 = refutes
# 2 = neutral
LABEL_NAMES = ["supports", "refutes", "neutral"]
LABEL2ID = {
    "supports": 0,
    "support": 0,
    "refutes": 1,
    "refute": 1,
    "neutral": 2,
}
ID2LABEL = {0: "supports", 1: "refutes", 2: "neutral"}

OUTPUT_DIR = "./outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


# ══════════════════════════════════════════════════════════════════════════════
# 1. Data Loading
# ══════════════════════════════════════════════════════════════════════════════
def load_healthver_csv(path):
    """
    Load a HealthVer CSV file with columns like:
    claim, evidence, label
    """
    df = pd.read_csv(path)

    required = {"claim", "evidence", "label"}
    missing = required - set(df.columns)
    if missing:
        raise ValueError(f"{path} is missing columns: {missing}")

    df = df.copy()
    df["claim"] = df["claim"].astype(str).str.strip()
    df["evidence"] = df["evidence"].astype(str).str.strip()
    df["label"] = (
        df["label"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map(LABEL2ID)
    )

    df = df.dropna(subset=["claim", "evidence", "label"]).copy()
    df["label"] = df["label"].astype(int)

    return df[["claim", "evidence", "label"]]


def split_data(df, val_size=0.2):
    """Stratified train/val split."""
    train_df, val_df = train_test_split(
        df, test_size=val_size, stratify=df["label"], random_state=SEED
    )
    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)
    print(f"\n  Train: {len(train_df)} | Val: {len(val_df)}")
    return train_df, val_df


# ══════════════════════════════════════════════════════════════════════════════
# 2. PyTorch Dataset
# ══════════════════════════════════════════════════════════════════════════════
class HealthVerDataset(Dataset):
    """
    Input to BERT: [CLS] claim [SEP] evidence [SEP]
    """
    def __init__(self, df: pd.DataFrame, tokenizer, max_len=MAX_LEN):
        self.df = df
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        claim = str(row["claim"])
        evid = str(row["evidence"])
        label = int(row["label"])

        enc = self.tokenizer(
            claim,
            evid,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_tensors="pt",
        )

        return {
            "input_ids": enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "token_type_ids": enc["token_type_ids"].squeeze(0),
            "label": torch.tensor(label, dtype=torch.long),
            "claim": claim,
            "evidence": evid,
        }


# ══════════════════════════════════════════════════════════════════════════════
# 3. Shortcut Feature Functions
# ══════════════════════════════════════════════════════════════════════════════
def lexical_overlap(claim: str, evidence: str) -> float:
    c_words = set(claim.lower().split())
    e_words = set(evidence.lower().split())
    if not e_words:
        return 0.0
    return len(c_words & e_words) / len(e_words)


def has_negation(text: str) -> int:
    neg_words = {
        "not", "no", "never", "neither", "nor", "n't",
        "nobody", "nothing", "none", "without", "lack", "lacking"
    }
    return int(bool(set(text.lower().split()) & neg_words))


def compute_shortcut_features(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out["overlap"] = out.apply(lambda r: lexical_overlap(r["claim"], r["evidence"]), axis=1)
    out["negation"] = out["evidence"].apply(has_negation)
    out["claim_len"] = out["claim"].apply(lambda x: len(x.split()))
    out["evid_len"] = out["evidence"].apply(lambda x: len(x.split()))
    out["label_name"] = out["label"].map(ID2LABEL)
    return out


# ══════════════════════════════════════════════════════════════════════════════
# 4. Training & Evaluation
# ══════════════════════════════════════════════════════════════════════════════
def train_epoch(model, loader, optimizer, scheduler, criterion, device):
    model.train()
    total_loss = 0.0
    correct = 0
    total = 0

    for batch in loader:
        ids = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        ttype = batch["token_type_ids"].to(device)
        labs = batch["label"].to(device)

        optimizer.zero_grad()
        logits = model(
            input_ids=ids,
            attention_mask=mask,
            token_type_ids=ttype
        ).logits
        loss = criterion(logits, labs)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()

        total_loss += loss.item()
        correct += (logits.argmax(-1) == labs).sum().item()
        total += labs.size(0)

    return total_loss / len(loader), correct / total


def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss = 0.0
    correct = 0
    total = 0
    all_preds, all_labels, all_probs = [], [], []

    with torch.no_grad():
        for batch in loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)
            labs = batch["label"].to(device)

            logits = model(
                input_ids=ids,
                attention_mask=mask,
                token_type_ids=ttype
            ).logits
            probs = torch.softmax(logits, dim=-1)
            loss = criterion(logits, labs)

            total_loss += loss.item()
            preds = logits.argmax(-1)
            correct += (preds == labs).sum().item()
            total += labs.size(0)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labs.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())

    return (
        total_loss / len(loader),
        correct / total,
        np.array(all_preds),
        np.array(all_labels),
        np.array(all_probs),
    )


def run_training(train_ds, val_ds, class_weights=None, tag="baseline"):
    """
    Fine-tune BERT on HealthVer.
    class_weights: None -> standard CE; torch.Tensor -> weighted CE
    """
    print(f"\n{'='*60}")
    print(f"  Training: {tag}")
    print(f"{'='*60}")

    model = BertForSequenceClassification.from_pretrained(
        "bert-base-uncased",
        num_labels=3
    ).to(DEVICE)

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    if class_weights is not None:
        criterion = nn.CrossEntropyLoss(weight=class_weights.to(DEVICE))
        print(f"  Class weights: {[f'{w:.3f}' for w in class_weights.tolist()]}")
    else:
        criterion = nn.CrossEntropyLoss()

    optimizer = AdamW(model.parameters(), lr=LR, weight_decay=0.01)
    total_steps = len(train_loader) * EPOCHS
    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=int(0.1 * total_steps),
        num_training_steps=total_steps,
    )

    history = []
    for epoch in range(1, EPOCHS + 1):
        tr_loss, tr_acc = train_epoch(
            model, train_loader, optimizer, scheduler, criterion, DEVICE
        )
        vl_loss, vl_acc, vl_preds, vl_labels, vl_probs = evaluate(
            model, val_loader, criterion, DEVICE
        )
        history.append({
            "epoch": epoch,
            "tr_loss": tr_loss,
            "tr_acc": tr_acc,
            "vl_loss": vl_loss,
            "vl_acc": vl_acc
        })
        print(
            f"  Epoch {epoch}/{EPOCHS} | "
            f"Train Loss {tr_loss:.4f}  Acc {tr_acc:.4f} | "
            f"Val Loss {vl_loss:.4f}  Acc {vl_acc:.4f}"
        )

    save_path = os.path.join(OUTPUT_DIR, f"bert_{tag}")
    model.save_pretrained(save_path)
    print(f"  Model saved → {save_path}")

    return model, vl_preds, vl_labels, vl_probs, history


# ══════════════════════════════════════════════════════════════════════════════
# 5. Temperature Calibration
# ══════════════════════════════════════════════════════════════════════════════
class TemperatureScaler(nn.Module):
    def __init__(self):
        super().__init__()
        self.temperature = nn.Parameter(torch.ones(1))

    def forward(self, logits):
        return logits / self.temperature

    def fit(self, model, val_loader, device, lr=0.01, max_iter=50):
        self.cpu()
        model.eval()

        logits_list, labels_list = [], []
        with torch.no_grad():
            for batch in val_loader:
                ids = batch["input_ids"].to(device)
                mask = batch["attention_mask"].to(device)
                ttype = batch["token_type_ids"].to(device)
                labs = batch["label"].to(device)

                out = model(
                    input_ids=ids,
                    attention_mask=mask,
                    token_type_ids=ttype
                )
                logits_list.append(out.logits.cpu())
                labels_list.append(labs.cpu())

        logits_all = torch.cat(logits_list)
        labels_all = torch.cat(labels_list)

        criterion = nn.CrossEntropyLoss()
        opt = torch.optim.LBFGS([self.temperature], lr=lr, max_iter=max_iter)

        def step():
            opt.zero_grad()
            loss = criterion(self.forward(logits_all), labels_all)
            loss.backward()
            return loss

        opt.step(step)
        T = self.temperature.item()
        print(f"  Learned temperature T = {T:.4f}  (>1 means model was overconfident)")

        cal_probs = torch.softmax(self.forward(logits_all).detach(), dim=-1).numpy()
        cal_preds = cal_probs.argmax(axis=-1)
        return cal_preds, labels_all.numpy(), cal_probs, T


# ══════════════════════════════════════════════════════════════════════════════
# 6. Shortcut Analysis
# ══════════════════════════════════════════════════════════════════════════════
def shortcut_analysis(model, val_loader, val_df, device, tag="baseline"):
    """
    Collect per-example: model confidence, lexical overlap, correctness.
    """
    model.eval()
    rows = []

    with torch.no_grad():
        for batch in val_loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)
            labs = batch["label"].to(device)
            claims = batch["claim"]
            evids = batch["evidence"]

            logits = model(
                input_ids=ids,
                attention_mask=mask,
                token_type_ids=ttype
            ).logits
            probs = torch.softmax(logits, dim=-1).cpu().numpy()
            preds = probs.argmax(axis=-1)
            labs_np = labs.cpu().numpy()

            for i in range(len(labs_np)):
                rows.append({
                    "label": labs_np[i],
                    "pred": preds[i],
                    "confidence": float(probs[i].max()),
                    "overlap": lexical_overlap(claims[i], evids[i]),
                    "correct": int(preds[i] == labs_np[i]),
                    "tag": tag,
                })

    df = pd.DataFrame(rows)
    df["label_name"] = df["label"].map(ID2LABEL)

    r, p = pearsonr(df["overlap"], df["confidence"])
    print(f"\n[{tag}] Overlap–Confidence Pearson r = {r:.4f}  (p = {p:.2e})")
    if r > 0.1:
        print("  ⚠ BERT confidence rises with lexical overlap → shortcut reliance!")
    else:
        print("  Weak overlap–confidence link")

    df["overlap_quartile"] = pd.qcut(
        df["overlap"],
        q=4,
        labels=["Q1 (low)", "Q2", "Q3", "Q4 (high)"]
    )
    print(f"\n[{tag}] Accuracy by lexical overlap quartile:")
    print(df.groupby("overlap_quartile")["correct"].mean().to_string())

    print(f"\n[{tag}] Accuracy by class:")
    print(df.groupby("label_name")["correct"].mean().to_string())

    return df, r


# ══════════════════════════════════════════════════════════════════════════════
# 7. Visualisation — 9-panel figure
# ══════════════════════════════════════════════════════════════════════════════
def plot_all(
    raw_df,
    base_df, base_preds, base_labels, base_probs,
    wt_df, wt_preds, wt_labels, wt_probs,
    cal_preds, cal_labels, cal_probs,
    base_r, wt_r,
):
    COLORS = ["steelblue", "tomato", "orange"]  # supports / refutes / neutral
    fig, axes = plt.subplots(3, 3, figsize=(18, 15))
    fig.suptitle(
        "BERT Shortcut Analysis on HealthVer\n"
        "Hypothesis: BERT exploits lexical overlap shortcuts instead of genuine semantic reasoning",
        fontsize=13,
        fontweight="bold",
    )

    # 1: Label distribution
    ax = axes[0, 0]
    counts = [Counter(raw_df["label"])[i] for i in range(3)]
    bars = ax.bar(LABEL_NAMES, counts, color=COLORS)
    for b, c in zip(bars, counts):
        ax.text(b.get_x() + b.get_width()/2, b.get_height() + 10, str(c), ha="center", fontsize=9)
    ax.set_title("Label Distribution")
    ax.set_ylabel("Count")
    ax.tick_params(axis="x", rotation=15)

    # 2: Lexical overlap distribution by label
    ax = axes[0, 1]
    for name, color in zip(LABEL_NAMES, COLORS):
        sub = raw_df[raw_df["label_name"] == name]["overlap"]
        ax.hist(sub, bins=30, alpha=0.6, label=name, color=color, density=True)
    ax.set_title("Lexical Overlap by Label")
    ax.set_xlabel("Overlap Score")
    ax.set_ylabel("Density")
    ax.legend(fontsize=8)

    # 3: Confidence vs overlap scatter (baseline)
    ax = axes[0, 2]
    for name, color in zip(LABEL_NAMES, COLORS):
        sub = base_df[base_df["label_name"] == name]
        ax.scatter(sub["overlap"], sub["confidence"], alpha=0.25, s=8, label=name, color=color)
    ax.set_title(f"Baseline: Confidence vs. Lexical Overlap\n(r = {base_r:.3f})")
    ax.set_xlabel("Lexical Overlap")
    ax.set_ylabel("Model Confidence")
    ax.legend(markerscale=3, fontsize=8)

    # 4: Accuracy by overlap quartile: baseline vs weighted
    ax = axes[1, 0]
    bq = base_df.groupby("overlap_quartile")["correct"].mean()
    wq = wt_df.groupby("overlap_quartile")["correct"].mean()
    x = np.arange(4)
    ax.bar(x - 0.2, bq.values, 0.4, label="Baseline", color="steelblue")
    ax.bar(x + 0.2, wq.values, 0.4, label="Class-Weighted", color="coral")
    ax.set_xticks(x)
    ax.set_xticklabels(["Q1\n(low overlap)", "Q2", "Q3", "Q4\n(high overlap)"])
    ax.set_title("Accuracy by Lexical Overlap Quartile")
    ax.set_ylabel("Accuracy")
    ax.legend()
    ax.set_ylim(0, 1)

    # 5: Confusion matrix — baseline
    ax = axes[1, 1]
    cm = confusion_matrix(base_labels, base_preds)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=ax,
                xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES)
    ax.set_title("Confusion Matrix: Baseline")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.tick_params(axis="x", rotation=20)

    # 6: Confusion matrix — class-weighted
    ax = axes[1, 2]
    cm_w = confusion_matrix(wt_labels, wt_preds)
    sns.heatmap(cm_w, annot=True, fmt="d", cmap="Oranges", ax=ax,
                xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES)
    ax.set_title("Confusion Matrix: Class-Weighted")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.tick_params(axis="x", rotation=20)

    # 7: Per-class F1 comparison
    ax = axes[2, 0]
    base_f1 = f1_score(base_labels, base_preds, average=None, labels=[0, 1, 2])
    wt_f1 = f1_score(wt_labels, wt_preds, average=None, labels=[0, 1, 2])
    cal_f1 = f1_score(cal_labels, cal_preds, average=None, labels=[0, 1, 2])
    x = np.arange(3)
    ax.bar(x - 0.27, base_f1, 0.27, label="Baseline", color="steelblue")
    ax.bar(x, wt_f1, 0.27, label="Class-Weighted", color="coral")
    ax.bar(x + 0.27, cal_f1, 0.27, label="Calibrated", color="mediumseagreen")
    ax.set_xticks(x)
    ax.set_xticklabels(LABEL_NAMES, rotation=10)
    ax.set_title("Per-Class F1")
    ax.set_ylabel("F1 Score")
    ax.legend(fontsize=8)
    ax.set_ylim(0, 1)

    # 8: Calibration curve — baseline
    ax = axes[2, 1]
    for cls_id, (name, color) in enumerate(zip(LABEL_NAMES, COLORS)):
        bl = (base_labels == cls_id).astype(int)
        try:
            fp, mp = calibration_curve(bl, base_probs[:, cls_id], n_bins=10)
            ax.plot(mp, fp, marker="o", label=name, color=color)
        except ValueError:
            pass
    ax.plot([0, 1], [0, 1], "k--", label="Perfect")
    ax.set_title("Calibration Curve: Baseline")
    ax.set_xlabel("Mean Predicted Probability")
    ax.set_ylabel("Fraction of Positives")
    ax.legend(fontsize=8)

    # 9: Calibration curve — after temperature scaling
    ax = axes[2, 2]
    for cls_id, (name, color) in enumerate(zip(LABEL_NAMES, COLORS)):
        bl = (cal_labels == cls_id).astype(int)
        try:
            fp, mp = calibration_curve(bl, cal_probs[:, cls_id], n_bins=10)
            ax.plot(mp, fp, marker="o", label=name, color=color)
        except ValueError:
            pass
    ax.plot([0, 1], [0, 1], "k--", label="Perfect")
    ax.set_title("Calibration: After Temperature Scaling")
    ax.set_xlabel("Mean Predicted Probability")
    ax.set_ylabel("Fraction of Positives")
    ax.legend(fontsize=8)

    plt.tight_layout()
    out = os.path.join(OUTPUT_DIR, "shortcut_analysis_healthver.png")
    plt.savefig(out, dpi=150, bbox_inches="tight")
    print(f"\nFigure saved → {out}")
    plt.show()


# ══════════════════════════════════════════════════════════════════════════════
# 8. Main
# ══════════════════════════════════════════════════════════════════════════════
def main():
    print(f"Device: {DEVICE}\n")

    train_path = "./data/healthver_train.csv"
    dev_path = "./data/healthver_dev.csv"
    test_path = "./data/healthver_test.csv"

    if not os.path.exists(train_path):
        raise FileNotFoundError(f"Missing file: {train_path}")
    if not os.path.exists(dev_path):
        raise FileNotFoundError(f"Missing file: {dev_path}")
    if not os.path.exists(test_path):
        raise FileNotFoundError(f"Missing file: {test_path}")

    train_df = load_healthver_csv(train_path)
    dev_df = load_healthver_csv(dev_path)
    test_df = load_healthver_csv(test_path)

    full_df = pd.concat([train_df, dev_df, test_df], ignore_index=True)

    print("HealthVer loaded:")
    print(f"  Train: {len(train_df)}")
    print(f"  Dev:   {len(dev_df)}")
    print(f"  Test:  {len(test_df)}")
    print("\nLabel distribution in train:")
    counts = Counter(train_df["label"])
    for i, name in enumerate(LABEL_NAMES):
        n = counts[i]
        print(f"  {name:10s}: {n:5d} ({n/len(train_df)*100:.1f}%)")

    tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
    train_ds = HealthVerDataset(train_df, tokenizer)
    dev_ds = HealthVerDataset(dev_df, tokenizer)
    test_ds = HealthVerDataset(test_df, tokenizer)

    dev_loader = DataLoader(dev_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    # Raw shortcut features on dev set
    print("\nComputing raw shortcut features on dev set...")
    raw_df = compute_shortcut_features(dev_df)
    print("  Mean lexical overlap by label:")
    print(raw_df.groupby("label_name")["overlap"].mean().to_string())
    print("\n  Negation rate by label:")
    print(raw_df.groupby("label_name")["negation"].mean().to_string())
    print("\n  Mean evidence length by label (tokens):")
    print(raw_df.groupby("label_name")["evid_len"].mean().to_string())

    # Class weights
    total = sum(counts.values())
    weights = torch.tensor(
        [total / (3 * counts[i]) for i in range(3)],
        dtype=torch.float
    )
    print(f"\nClass weights (inv-freq): {[f'{w:.3f}' for w in weights.tolist()]}")

    # Baseline BERT
    base_model, base_preds, base_labels, base_probs, _ = run_training(
        train_ds, dev_ds, class_weights=None, tag="baseline"
    )

    # Class-weighted BERT
    wt_model, wt_preds, wt_labels, wt_probs, _ = run_training(
        train_ds, dev_ds, class_weights=weights, tag="class_weighted"
    )

    # Temperature calibration on baseline
    print("\nFitting temperature scaling (post-hoc calibration)...")
    scaler = TemperatureScaler()
    cal_preds, cal_labels, cal_probs, T = scaler.fit(base_model, dev_loader, DEVICE)

    # Shortcut analysis
    print("\nShortcut analysis — baseline model:")
    base_df, base_r = shortcut_analysis(base_model, dev_loader, dev_df, DEVICE, "baseline")
    print("\nShortcut analysis — class-weighted model:")
    wt_df, wt_r = shortcut_analysis(wt_model, dev_loader, dev_df, DEVICE, "class_weighted")

    # Dev reports
    for tag, labs, preds in [
        ("Baseline", base_labels, base_preds),
        ("Class-Weighted", wt_labels, wt_preds),
        ("Baseline + Calibration", cal_labels, cal_preds),
    ]:
        print(f"\n{'─'*55}")
        print(f"  {tag}")
        print(f"  Overall Accuracy: {accuracy_score(labs, preds):.4f}")
        print(classification_report(labs, preds, target_names=LABEL_NAMES))

    # Optional final test evaluation on the baseline model
    print("\nFinal evaluation on test set (baseline model):")
    criterion = nn.CrossEntropyLoss()
    test_loss, test_acc, test_preds, test_labels, test_probs = evaluate(
        base_model, test_loader, criterion, DEVICE
    )
    print(f"  Test Loss: {test_loss:.4f}")
    print(f"  Test Acc : {test_acc:.4f}")
    print(classification_report(test_labels, test_preds, target_names=LABEL_NAMES))

    # Save JSON summary
    results = {
        "dataset": "HealthVer",
        "model": "bert-base-uncased",
        "train_n": len(train_df),
        "dev_n": len(dev_df),
        "test_n": len(test_df),
        "baseline": {
            "dev_accuracy": accuracy_score(base_labels, base_preds),
            "dev_f1_per_class": f1_score(base_labels, base_preds, average=None, labels=[0, 1, 2]).tolist(),
            "overlap_confidence_pearsonr": base_r,
            "test_accuracy": test_acc,
            "test_f1_per_class": f1_score(test_labels, test_preds, average=None, labels=[0, 1, 2]).tolist(),
        },
        "class_weighted": {
            "dev_accuracy": accuracy_score(wt_labels, wt_preds),
            "dev_f1_per_class": f1_score(wt_labels, wt_preds, average=None, labels=[0, 1, 2]).tolist(),
            "overlap_confidence_pearsonr": wt_r,
        },
        "calibration": {
            "learned_temperature": T,
            "dev_accuracy": accuracy_score(cal_labels, cal_preds),
            "dev_f1_per_class": f1_score(cal_labels, cal_preds, average=None, labels=[0, 1, 2]).tolist(),
        },
    }

    json_path = os.path.join(OUTPUT_DIR, "results_summary_healthver.json")
    with open(json_path, "w") as f:
        json.dump(results, f, indent=2)
    print(f"\nResults JSON → {json_path}")

    # Plots
    plot_all(
        raw_df,
        base_df, base_preds, base_labels, base_probs,
        wt_df, wt_preds, wt_labels, wt_probs,
        cal_preds, cal_labels, cal_probs,
        base_r, wt_r,
    )

    print("\n✓ All done. Outputs saved to ./outputs/")


if __name__ == "__main__":
    main()
```

Synonym-Swap

```python
def get_synonym(word):
    syns = set()
    for syn in wn.synsets(word):
        for lemma in syn.lemmas():
            cand = lemma.name().replace("_", " ")
            if cand.lower() != word.lower():
                syns.add(cand)
    return sorted(syns)[0] if syns else None

def synonym_swap(text, p=0.2):
    tokens = word_tokenize(text)
    out = []
    for tok in tokens:
        if re.fullmatch(r"[A-Za-z]+", tok) and len(tok) > 3 and random.random() < p:
            syn = get_synonym(tok)
            out.append(syn if syn else tok)
        else:
            out.append(tok)
    return " ".join(out)

def run_synonym_swap_test(model, df, tokenizer, device, max_len=MAX_LEN, p=0.2):
    """
    Returns confidence drop and accuracy delta after synonym swapping.
    """
    pert = df.copy()
    pert["claim"] = pert["claim"].apply(lambda x: synonym_swap(x, p=p))
    pert["evidence"] = pert["evidence"].apply(lambda x: synonym_swap(x, p=p))

    orig_ds = HealthVerDataset(df, tokenizer, max_len=max_len)
    pert_ds = HealthVerDataset(pert, tokenizer, max_len=max_len)
    orig_loader = DataLoader(orig_ds, batch_size=BATCH_SIZE, shuffle=False)
    pert_loader = DataLoader(pert_ds, batch_size=BATCH_SIZE, shuffle=False)

    criterion = nn.CrossEntropyLoss()
    _, orig_acc, _, _, orig_probs = evaluate(model, orig_loader, criterion, device)
    _, pert_acc, _, _, pert_probs = evaluate(model, pert_loader, criterion, device)

    orig_conf = orig_probs.max(axis=1).mean()
    pert_conf = pert_probs.max(axis=1).mean()

    return {
        "orig_acc": orig_acc,
        "pert_acc": pert_acc,
        "acc_delta": pert_acc - orig_acc,
        "orig_conf": orig_conf,
        "pert_conf": pert_conf,
        "conf_drop": pert_conf - orig_conf,
    }
```

Attention Masking

```python
def overlap_words(claim, evidence):
    c = set(re.findall(r"\w+", claim.lower()))
    e = re.findall(r"\w+", evidence.lower())
    return set(c), e

def mask_overlap_in_evidence(claim, evidence):
    cset, e_tokens = overlap_words(claim, evidence)
    masked = []
    for tok in e_tokens:
        if tok.lower() in cset:
            masked.append("[MASK]")
        else:
            masked.append(tok)
    return " ".join(masked)

def run_attention_masking_test(model, df, tokenizer, device, max_len=MAX_LEN):
    """
    Masks overlapping evidence tokens and measures accuracy delta.
    """
    pert = df.copy()
    pert["evidence"] = pert.apply(
        lambda r: mask_overlap_in_evidence(r["claim"], r["evidence"]),
        axis=1
    )

    orig_ds = HealthVerDataset(df, tokenizer, max_len=max_len)
    pert_ds = HealthVerDataset(pert, tokenizer, max_len=max_len)
    orig_loader = DataLoader(orig_ds, batch_size=BATCH_SIZE, shuffle=False)
    pert_loader = DataLoader(pert_ds, batch_size=BATCH_SIZE, shuffle=False)

    criterion = nn.CrossEntropyLoss()
    _, orig_acc, _, _, _ = evaluate(model, orig_loader, criterion, device)
    _, pert_acc, _, _, _ = evaluate(model, pert_loader, criterion, device)

    return {
        "orig_acc": orig_acc,
        "masked_acc": pert_acc,
        "acc_delta": pert_acc - orig_acc,
    }
```

Probe (Overlap → CLS)

```python
class BertCLSExtractor(nn.Module):
    def __init__(self, bert_model):
        super().__init__()
        self.bert = bert_model.bert

    def forward(self, input_ids, attention_mask, token_type_ids):
        out = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids,
            return_dict=True
        )
        return out.pooler_output  # CLS-pooled representation

def extract_cls_embeddings(model, loader, device):
    model.eval()
    feats, ys = [], []
    with torch.no_grad():
        for batch in loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)

            out = model(
                input_ids=ids,
                attention_mask=mask,
                token_type_ids=ttype,
                return_dict=True
            )
            cls = out.pooler_output.detach().cpu().numpy()
            feats.append(cls)
    return np.concatenate(feats, axis=0)

def run_overlap_cls_probe(model, df, tokenizer, device, max_len=MAX_LEN):
    """
    Train a linear probe to predict high-vs-low overlap from CLS embeddings.
    Chance is ~0.5 if the split is balanced.
    """
    probe_df = compute_shortcut_features(df)
    threshold = probe_df["overlap"].median()
    probe_df["overlap_bin"] = (probe_df["overlap"] >= threshold).astype(int)

    ds = HealthVerDataset(probe_df, tokenizer, max_len=max_len)
    loader = DataLoader(ds, batch_size=BATCH_SIZE, shuffle=False)

    # Get CLS embeddings
    model.eval()
    X, y = [], []
    with torch.no_grad():
        for batch in loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)

            out = model(
                input_ids=ids,
                attention_mask=mask,
                token_type_ids=ttype,
                output_hidden_states=True,
                return_dict=True
            )
            cls = out.hidden_states[-1][:, 0, :].detach().cpu().numpy()
            X.append(cls)
            y.append(batch["label"].numpy())  # replace with overlap_bin below

    X = np.concatenate(X, axis=0)
    y = probe_df["overlap_bin"].values

    idx = np.arange(len(y))
    tr_idx, te_idx = train_test_split(
        idx, test_size=0.3, random_state=SEED, stratify=y
    )

    clf = make_pipeline(
        StandardScaler(),
        LogisticRegression(max_iter=1000)
    )
    clf.fit(X[tr_idx], y[tr_idx])
    pred = clf.predict(X[te_idx])
    acc = accuracy_score(y[te_idx], pred)
    bacc = balanced_accuracy_score(y[te_idx], pred)

    return {
        "probe_acc": acc,
        "probe_balanced_acc": bacc,
        "chance": 0.5,
        "threshold": float(threshold),
    }
```

NEI Deep-Dive

```python
def nei_deep_dive(y_true, y_pred, nei_label=2):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    nei_mask = (y_true == nei_label)
    non_nei_mask = ~nei_mask

    nei_acc = accuracy_score(y_true[nei_mask], y_pred[nei_mask]) if nei_mask.any() else np.nan
    non_nei_acc = accuracy_score(y_true[non_nei_mask], y_pred[non_nei_mask]) if non_nei_mask.any() else np.nan

    return {
        "nei_acc": nei_acc,
        "non_nei_acc": non_nei_acc,
        "gap": nei_acc - non_nei_acc,
        "nei_rate": float(nei_mask.mean()),
    }
```

Error Analysis

```python
def error_analysis_shortcuts(df, y_true, y_pred):
    tmp = df.copy()
    tmp["true"] = y_true
    tmp["pred"] = y_pred
    tmp["correct"] = (tmp["true"] == tmp["pred"]).astype(int)

    errors = tmp[tmp["correct"] == 0].copy()

    def shortcut_type(row):
        # Main bias: predicting NEI/neutral too often
        if row["pred"] == 2 and row["true"] != 2:
            return "label_bias_NEI"
        # Overlap-sensitive mistake
        if row["overlap"] > tmp["overlap"].median():
            return "high_overlap_failure"
        # Negation-sensitive mistake
        if row["negation"] == 1:
            return "negation_failure"
        return "other"

    errors["shortcut"] = errors.apply(shortcut_type, axis=1)
    counts = errors["shortcut"].value_counts()

    dominant = counts.index[0] if len(counts) else None
    dominant_n = int(counts.iloc[0]) if len(counts) else 0
    total_err = len(errors)

    return {
        "total_errors": total_err,
        "dominant_shortcut": dominant,
        "dominant_count": dominant_n,
        "dominant_share": dominant_n / total_err if total_err else np.nan,
        "counts": counts.to_dict(),
        "errors_df": errors,
    }
```
