```python
import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
from datasets import load_dataset

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from torch.optim import AdamW

from transformers import (
    BertTokenizer,
    BertForSequenceClassification,
    get_linear_schedule_with_warmup,
)
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score,
)
from sklearn.calibration import calibration_curve
from sklearn.model_selection import train_test_split
from scipy.stats import pearsonr

# --- Config -------------------------------------------------------------------
SEED       = 42
MAX_LEN    = 128
BATCH_SIZE = 16
EPOCHS     = 4
LR         = 2e-5
DEVICE     = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# PubHealth label mapping
# 0 = supports
# 1 = refutes
# 2 = neutral

LABEL_NAMES = ["supports", "refutes", "neutral"]

LABEL2ID = {
    "true": 0,
    "supports": 0,
    "support": 0,

    "false": 1,
    "refutes": 1,
    "refute": 1,

    "mixture": 2,
    "unproven": 2,
    "neutral": 2,
}

ID2LABEL = {0: "supports", 1: "refutes", 2: "neutral"}

ID2LABEL = {0: "supports", 1: "refutes", 2: "neutral"}

OUTPUT_DIR = "./outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


# ==============================================================================
# 1. Data Loading
# ==============================================================================
def normalize_pubhealth_label(label):
    label = str(label).strip().lower()
    if label == "true":
        return 0      # supports
    if label == "false":
        return 1      # refutes
    if label in {"mixture", "unproven"}:
        return 2      # neutral
    return None


def convert_pubhealth_split(df):
    out = df.copy()

    out["claim"] = out["text_1"].astype(str).str.strip()
    out["evidence"] = out["text_2"].astype(str).str.strip()
    out["label"] = out["label"].apply(normalize_pubhealth_label)

    out = out.dropna(subset=["claim", "evidence", "label"]).copy()
    out["label"] = out["label"].astype(int)

    return out[["claim", "evidence", "label"]]

def split_data(df, val_size=0.2):
    """Stratified train/val split."""
    train_df, val_df = train_test_split(
        df, test_size=val_size, stratify=df["label"], random_state=SEED
    )
    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)
    print(f"\n  Train: {len(train_df)} | Val: {len(val_df)}")
    return train_df, val_df


# ==============================================================================
# 2. PyTorch Dataset
# ==============================================================================
class PubHealthDataset(Dataset):
    """
    Input to BERT: [CLS] claim [SEP] evidence [SEP]
    """
    def __init__(self, df: pd.DataFrame, tokenizer, max_len=MAX_LEN):
        self.df = df
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        claim = str(row["claim"])
        evid = str(row["evidence"])
        label = int(row["label"])

        enc = self.tokenizer(
            claim,
            evid,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_tensors="pt",
        )

        return {
            "input_ids": enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "token_type_ids": enc["token_type_ids"].squeeze(0),
            "label": torch.tensor(label, dtype=torch.long),
            "claim": claim,
            "evidence": evid,
        }


# ==============================================================================
# 3. Shortcut Feature Functions
# ==============================================================================
def lexical_overlap(claim: str, evidence: str) -> float:
    c_words = set(claim.lower().split())
    e_words = set(evidence.lower().split())
    if not e_words:
        return 0.0
    return len(c_words & e_words) / len(e_words)


def has_negation(text: str) -> int:
    neg_words = {
        "not", "no", "never", "neither", "nor", "n't",
        "nobody", "nothing", "none", "without", "lack", "lacking"
    }
    return int(bool(set(text.lower().split()) & neg_words))


def compute_shortcut_features(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out["overlap"] = out.apply(lambda r: lexical_overlap(r["claim"], r["evidence"]), axis=1)
    out["negation"] = out["evidence"].apply(has_negation)
    out["claim_len"] = out["claim"].apply(lambda x: len(x.split()))
    out["evid_len"] = out["evidence"].apply(lambda x: len(x.split()))
    out["label_name"] = out["label"].map(ID2LABEL)
    return out


# ==============================================================================
# 4. Training & Evaluation
# ==============================================================================
def train_epoch(model, loader, optimizer, scheduler, criterion, device):
    model.train()
    total_loss = 0.0
    correct = 0
    total = 0

    for batch in loader:
        ids = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        ttype = batch["token_type_ids"].to(device)
        labs = batch["label"].to(device)

        optimizer.zero_grad()
        logits = model(
            input_ids=ids,
            attention_mask=mask,
            token_type_ids=ttype
        ).logits
        loss = criterion(logits, labs)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()

        total_loss += loss.item()
        correct += (logits.argmax(-1) == labs).sum().item()
        total += labs.size(0)

    return total_loss / len(loader), correct / total


def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss = 0.0
    correct = 0
    total = 0
    all_preds, all_labels, all_probs = [], [], []

    with torch.no_grad():
        for batch in loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)
            labs = batch["label"].to(device)

            logits = model(
                input_ids=ids,
                attention_mask=mask,
                token_type_ids=ttype
            ).logits
            probs = torch.softmax(logits, dim=-1)
            loss = criterion(logits, labs)

            total_loss += loss.item()
            preds = logits.argmax(-1)
            correct += (preds == labs).sum().item()
            total += labs.size(0)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labs.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())

    return (
        total_loss / len(loader),
        correct / total,
        np.array(all_preds),
        np.array(all_labels),
        np.array(all_probs),
    )


def run_training(train_ds, val_ds, class_weights=None, tag="baseline"):
    """
    Fine-tune BERT on PubHealth.
    class_weights: None -> standard CE; torch.Tensor -> weighted CE
    """
    print(f"\n{'='*60}")
    print(f"  Training: {tag}")
    print(f"{'='*60}")

    model = BertForSequenceClassification.from_pretrained(
        "bert-base-uncased",
        num_labels=3
    ).to(DEVICE)

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    if class_weights is not None:
        criterion = nn.CrossEntropyLoss(weight=class_weights.to(DEVICE))
        print(f"  Class weights: {[f'{w:.3f}' for w in class_weights.tolist()]}")
    else:
        criterion = nn.CrossEntropyLoss()

    optimizer = AdamW(model.parameters(), lr=LR, weight_decay=0.01)
    total_steps = len(train_loader) * EPOCHS
    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=int(0.1 * total_steps),
        num_training_steps=total_steps,
    )

    history = []
    for epoch in range(1, EPOCHS + 1):
        tr_loss, tr_acc = train_epoch(
            model, train_loader, optimizer, scheduler, criterion, DEVICE
        )
        vl_loss, vl_acc, vl_preds, vl_labels, vl_probs = evaluate(
            model, val_loader, criterion, DEVICE
        )
        history.append({
            "epoch": epoch,
            "tr_loss": tr_loss,
            "tr_acc": tr_acc,
            "vl_loss": vl_loss,
            "vl_acc": vl_acc
        })
        print(
            f"  Epoch {epoch}/{EPOCHS} | "
            f"Train Loss {tr_loss:.4f}  Acc {tr_acc:.4f} | "
            f"Val Loss {vl_loss:.4f}  Acc {vl_acc:.4f}"
        )

    save_path = os.path.join(OUTPUT_DIR, f"bert_{tag}")
    model.save_pretrained(save_path)
    print(f"  Model saved \u2192 {save_path}")

    return model, vl_preds, vl_labels, vl_probs, history


# ==============================================================================
# 5. Temperature Calibration
# ==============================================================================
class TemperatureScaler(nn.Module):
    def __init__(self):
        super().__init__()
        self.temperature = nn.Parameter(torch.ones(1))

    def forward(self, logits):
        return logits / self.temperature

    def fit(self, model, val_loader, device, lr=0.01, max_iter=50):
        self.cpu()
        model.eval()

        logits_list, labels_list = [], []
        with torch.no_grad():
          for batch in val_loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)
            labs = batch["label"].to(device)

            out = model(
                  input_ids=ids,
                  attention_mask=mask,
                  token_type_ids=ttype
              ).logits

            logits_list.append(out.cpu())
            labels_list.append(labs.cpu())

        logits_all = torch.cat(logits_list)
        labels_all = torch.cat(labels_list)

        criterion = nn.CrossEntropyLoss()
        opt = torch.optim.LBFGS([self.temperature], lr=lr, max_iter=50)

        def step():
            opt.zero_grad()
            loss = criterion(self.forward(logits_all), labels_all)
            loss.backward()
            return loss

        opt.step(step)
        T = self.temperature.item()
        print(f"  Learned temperature T = {T:.4f}  (>1 means model was overconfident)")

        cal_probs = torch.softmax(self.forward(logits_all).detach(), dim=-1).numpy()
        cal_preds = cal_probs.argmax(axis=-1)
        return cal_preds, labels_all.numpy(), cal_probs, T


# ==============================================================================
# 6. Shortcut Analysis
# ==============================================================================
def shortcut_analysis(model, val_loader, val_df, device, tag="baseline"):
    """
    Collect per-example: model confidence, lexical overlap, correctness.
    """
    model.eval()
    rows = []

    with torch.no_grad():
        for batch in val_loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)
            labs = batch["label"].to(device)
            claims = batch["claim"]
            evids = batch["evidence"]

            logits = model(
                input_ids=ids,
                attention_mask=mask,
                token_type_ids=ttype
            ).logits
            probs = torch.softmax(logits, dim=-1).cpu().numpy()
            preds = probs.argmax(axis=-1)
            labs_np = labs.cpu().numpy()

            for i in range(len(labs_np)):
                rows.append({
                    "label": labs_np[i],
                    "pred": preds[i],
                    "confidence": float(probs[i].max()),
                    "overlap": lexical_overlap(claims[i], evids[i]),
                    "correct": int(preds[i] == labs_np[i]),
                    "tag": tag,
                })

    df = pd.DataFrame(rows)
    df["label_name"] = df["label"].map(ID2LABEL)

    if df["overlap"].nunique() > 1 and df["confidence"].nunique() > 1:
      r, p = pearsonr(df["overlap"], df["confidence"])
    else:
      r, p = np.nan, np.nan

    print(f"[{tag}] Overlap\u2013Confidence Pearson r = {r}  (p = {p})")
    if r > 0.1:
        print("  \u26a0 BERT confidence rises with lexical overlap \u2192 shortcut reliance!")
    else:
        print("  Weak overlap\u2013confidence link")

    # Safe quartiles even when overlap has repeated values
    if df["overlap"].nunique() >= 4:
        df["overlap_quartile"] = pd.qcut(
            df["overlap"],
            q=4,
            labels=["Q1 (low)", "Q2", "Q3", "Q4 (high)"],
            duplicates="drop",
        )
    else:
        # Fallback: put everything into one bucket if overlap is constant
        df["overlap_quartile"] = "Q1 (low)"
    print(f"\n[{tag}] Accuracy by lexical overlap quartile:")
    print(df.groupby("overlap_quartile")["correct"].mean().to_string())

    print(f"\n[{tag}] Accuracy by class:")
    print(df.groupby("label_name")["correct"].mean().to_string())

    return df, r


# ==============================================================================
# 7. Visualisation \u2014 9-panel figure
# ==============================================================================
def plot_all(
    raw_df,
    base_df, base_preds, base_labels, base_probs,
    wt_df, wt_preds, wt_labels, wt_probs,
    cal_preds, cal_labels, cal_probs,
    base_r, wt_r,
):
    COLORS = ["steelblue", "tomato", "orange"]  # supports / refutes / neutral
    fig, axes = plt.subplots(3, 3, figsize=(18, 15))
    fig.suptitle(
        "BERT Shortcut Analysis on PubHealth\n"
        "Hypothesis: BERT exploits lexical overlap shortcuts instead of genuine semantic reasoning",
        fontsize=13,
        fontweight="bold",
    )

    # 1: Label distribution
    ax = axes[0, 0]
    counts = [Counter(raw_df["label"])[i] for i in range(3)]
    bars = ax.bar(LABEL_NAMES, counts, color=COLORS)
    for b, c in zip(bars, counts):
        ax.text(b.get_x() + b.get_width()/2, b.get_height() + 10, str(c), ha="center", fontsize=9)
    ax.set_title("Label Distribution")
    ax.set_ylabel("Count")
    ax.tick_params(axis="x", rotation=15)

    # 2: Lexical overlap distribution by label
    ax = axes[0, 1]
    for name, color in zip(LABEL_NAMES, COLORS):
        sub = raw_df[raw_df["label_name"] == name]["overlap"]
        ax.hist(sub, bins=30, alpha=0.6, label=name, color=color, density=True)
    ax.set_title("Lexical Overlap by Label")
    ax.set_xlabel("Overlap Score")
    ax.set_ylabel("Density")
    ax.legend(fontsize=8)

    # 3: Confidence vs overlap scatter (baseline)
    ax = axes[0, 2]
    for name, color in zip(LABEL_NAMES, COLORS):
        sub = base_df[base_df["label_name"] == name]
        ax.scatter(sub["overlap"], sub["confidence"], alpha=0.25, s=8, label=name, color=color)
    ax.set_title(f"Baseline: Confidence vs. Lexical Overlap\n(r = {base_r:.3f})")
    ax.set_xlabel("Lexical Overlap")
    ax.set_ylabel("Model Confidence")
    ax.legend(markerscale=3, fontsize=8)

    # 4: Accuracy by overlap quartile: baseline vs weighted
    ax = axes[1, 0]
    bq = base_df.groupby("overlap_quartile")["correct"].mean()
    wq = wt_df.groupby("overlap_quartile")["correct"].mean()
    x = np.arange(4)
    ax.bar(x - 0.2, bq.values, 0.4, label="Baseline", color="steelblue")
    ax.bar(x + 0.2, wq.values, 0.4, label="Class-Weighted", color="coral")
    ax.set_xticks(x)
    ax.set_xticklabels(["Q1\n(low overlap)", "Q2", "Q3", "Q4\n(high overlap)"])
    ax.set_title("Accuracy by Lexical Overlap Quartile")
    ax.set_ylabel("Accuracy")
    ax.legend()
    ax.set_ylim(0, 1)

    # 5: Confusion matrix \u2014 baseline
    ax = axes[1, 1]
    cm = confusion_matrix(base_labels, base_preds)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=ax,
                xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES)
    ax.set_title("Confusion Matrix: Baseline")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.tick_params(axis="x", rotation=20)

    # 6: Confusion matrix \u2014 class-weighted
    ax = axes[1, 2]
    cm_w = confusion_matrix(wt_labels, wt_preds)
    sns.heatmap(cm_w, annot=True, fmt="d", cmap="Oranges", ax=ax,
                xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES)
    ax.set_title("Confusion Matrix: Class-Weighted")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.tick_params(axis="x", rotation=20)

    # 7: Per-class F1 comparison
    ax = axes[2, 0]
    base_f1 = f1_score(base_labels, base_preds, average=None, labels=[0, 1, 2])
    wt_f1 = f1_score(wt_labels, wt_preds, average=None, labels=[0, 1, 2])
    cal_f1 = f1_score(cal_labels, cal_preds, average=None, labels=[0, 1, 2])
    x = np.arange(3)
    ax.bar(x - 0.27, base_f1, 0.27, label="Baseline", color="steelblue")
    ax.bar(x, wt_f1, 0.27, label="Class-Weighted", color="coral")
    ax.bar(x + 0.27, cal_f1, 0.27, label="Calibrated", color="mediumseagreen")
    ax.set_xticks(x)
    ax.set_xticklabels(LABEL_NAMES, rotation=10)
    ax.set_title("Per-Class F1")
    ax.set_ylabel("F1 Score")
    ax.legend(fontsize=8)
    ax.set_ylim(0, 1)

    # 8: Calibration curve \u2014 baseline
    ax = axes[2, 1]
    for cls_id, (name, color) in enumerate(zip(LABEL_NAMES, COLORS)):
        bl = (base_labels == cls_id).astype(int)
        try:
            fp, mp = calibration_curve(bl, base_probs[:, cls_id], n_bins=10)
            ax.plot(mp, fp, marker="o", label=name, color=color)
        except ValueError:
            pass
    ax.plot([0, 1], [0, 1], "k--", label="Perfect")
    ax.set_title("Calibration Curve: Baseline")
    ax.set_xlabel("Mean Predicted Probability")
    ax.set_ylabel("Fraction of Positives")
    ax.legend(fontsize=8)

    # 9: Calibration curve \u2014 after temperature scaling
    ax = axes[2, 2]
    for cls_id, (name, color) in enumerate(zip(LABEL_NAMES, COLORS)):
        bl = (cal_labels == cls_id).astype(int)
        try:
            fp, mp = calibration_curve(bl, cal_probs[:, cls_id], n_bins=10)
            ax.plot(mp, fp, marker="o", label=name, color=color)
        except ValueError:
            pass
    ax.plot([0, 1], [0, 1], "k--", label="Perfect")
    ax.set_title("Calibration: After Temperature Scaling")
    ax.set_xlabel("Mean Predicted Probability")
    ax.set_ylabel("Fraction of Positives")
    ax.legend(fontsize=8)

    plt.tight_layout()
    out = os.path.join(OUTPUT_DIR, "shortcut_analysis_pubhealth.png")
    plt.savefig(out, dpi=150, bbox_inches="tight")
    print(f"\nFigure saved \u2192 {out}")
    plt.show()


# ==============================================================================
# 8. Main
# ==============================================================================
def main():
    print(f"Device: {DEVICE}\n")

    dataset = load_dataset("bigbio/pubhealth", "pubhealth_bigbio_pairs")

    train_df = convert_pubhealth_split(dataset["train"].to_pandas())
    dev_df   = convert_pubhealth_split(dataset["validation"].to_pandas())
    test_df  = convert_pubhealth_split(dataset["test"].to_pandas())
    for name, df_ in [("Train", train_df), ("Dev", dev_df), ("Test", test_df)]:
      print(f"\n{name} label distribution:")
      print(df_["label"].value_counts().sort_index())

    print("PubHealth loaded:")
    print(f"  Train: {len(train_df)}")
    print(f"  Dev:   {len(dev_df)}")
    print(f"  Test:  {len(test_df)}")
    print("\nLabel distribution in train:")
    counts = Counter(train_df["label"])
    # Modified weight calculation to handle classes with zero samples
    weights_list = []
    for i in range(len(LABEL_NAMES)): # Iterate through all potential labels (0, 1, 2)
        n = counts[i]
        print(f"  {LABEL_NAMES[i]:10s}: {n:5d} ({n/len(train_df)*100:.1f}%)")
        if n > 0:
            # Calculate inverse frequency weight, scaled by the total number of labels (3)
            weights_list.append(len(train_df) / (len(LABEL_NAMES) * n))
        else:
            # Assign 0 weight for non-existent classes
            weights_list.append(0.0)

    tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
    train_ds = PubHealthDataset(train_df, tokenizer)
    dev_ds = PubHealthDataset(dev_df, tokenizer)
    test_ds = PubHealthDataset(test_df, tokenizer)

    dev_loader = DataLoader(dev_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    # Raw shortcut features on dev set
    print("\nComputing raw shortcut features on dev set...")
    raw_df = compute_shortcut_features(dev_df)

    print("  Mean lexical overlap by label:")
    print(raw_df.groupby("label_name")["overlap"].mean().to_string())

    nei_raw = raw_df[raw_df["label"] == 2]
    print("\nNEI-only stats on dev:")
    print(f"  Count: {len(nei_raw)}")
    if len(nei_raw) > 0:
      print(f"  Mean overlap: {nei_raw['overlap'].mean():.4f}")
      print(f"  Mean evidence length: {nei_raw['evid_len'].mean():.4f}")
    print("\n  Negation rate by label:")
    print(raw_df.groupby("label_name")["negation"].mean().to_string())
    print("\n  Mean evidence length by label (tokens):")
    print(raw_df.groupby("label_name")["evid_len"].mean().to_string())

    # Class weights
    weights = torch.tensor(weights_list, dtype=torch.float)
    print(f"\nClass weights (inv-freq): {[f'{w:.3f}' for w in weights.tolist()]}")

    # Baseline BERT
    base_model, base_preds, base_labels, base_probs, _ = run_training(
        train_ds, dev_ds, class_weights=None, tag="baseline"
    )

    # Class-weighted BERT
    wt_model, wt_preds, wt_labels, wt_probs, _ = run_training(
        train_ds, dev_ds, class_weights=weights, tag="class_weighted"
    )

    # Temperature calibration on baseline
    print("\nFitting temperature scaling (post-hoc calibration)...")
    scaler = TemperatureScaler()
    cal_preds, cal_labels, cal_probs, T = scaler.fit(base_model, dev_loader, DEVICE)

    # Shortcut analysis
    print("\nShortcut analysis \u2014 baseline model:")
    base_df, base_r = shortcut_analysis(base_model, dev_loader, dev_df, DEVICE, "baseline")
    print("\nShortcut analysis \u2014 class-weighted model:")
    wt_df, wt_r = shortcut_analysis(wt_model, dev_loader, dev_df, DEVICE, "class_weighted")

    # Dev reports
    for tag, labs, preds in [
        ("Baseline", base_labels, base_preds),
        ("Class-Weighted", wt_labels, wt_preds),
        ("Baseline + Calibration", cal_labels, cal_preds),
    ]:
        print(f"\n{'\u2500'*55}")
        print(f"  {tag}")
        print(f"  Overall Accuracy: {accuracy_score(labs, preds):.4f}")
        print(classification_report(labs, preds, labels=[0, 1, 2], target_names=LABEL_NAMES, zero_division=0))

    # Final test evaluation on the baseline model
    test_acc = None
    test_f1_per_class = None
    if test_loader:
        print("\nFinal evaluation on test set (baseline model):")
        criterion = nn.CrossEntropyLoss()
        test_loss, test_acc, test_preds, test_labels, test_probs = evaluate(
            base_model, test_loader, criterion, DEVICE
        )
        print(f"  Test Loss: {test_loss:.4f}")
        print(f"  Test Acc : {test_acc:.4f}")
        print(classification_report(test_labels, test_preds, labels=[0, 1, 2], target_names=LABEL_NAMES, zero_division=0))
        test_f1_per_class = f1_score(test_labels, test_preds, average=None, labels=[0, 1, 2]).tolist()


    # Save JSON summary
    results = {
        "dataset": "PubHealth",
        "model": "bert-base-uncased",
        "train_n": len(train_df),
        "dev_n": len(dev_df),
        "test_n": len(test_df),
        "baseline": {
            "dev_accuracy": accuracy_score(base_labels, base_preds),
            "dev_f1_per_class": f1_score(base_labels, base_preds, average=None, labels=[0, 1, 2]).tolist(),
            "overlap_confidence_pearsonr": base_r,
            "test_accuracy": test_acc,
            "test_f1_per_class": test_f1_per_class,
        },
        "class_weighted": {
            "dev_accuracy": accuracy_score(wt_labels, wt_preds),
            "dev_f1_per_class": f1_score(wt_labels, wt_preds, average=None, labels=[0, 1, 2]).tolist(),
            "overlap_confidence_pearsonr": wt_r,
        },
        "calibration": {
            "learned_temperature": T,
            "dev_accuracy": accuracy_score(cal_labels, cal_preds),
            "dev_f1_per_class": f1_score(cal_labels, cal_preds, average=None, labels=[0, 1, 2]).tolist(),
        },
    }

    json_path = os.path.join(OUTPUT_DIR, "results_summary_pubhealth.json")
    with open(json_path, "w") as f:
        json.dump(results, f, indent=2)
    print(f"\nResults JSON \u2192 {json_path}")

    # Plots
    plot_all(
        raw_df,
        base_df, base_preds, base_labels, base_probs,
        wt_df, wt_preds, wt_labels, wt_probs,
        cal_preds, cal_labels, cal_probs,
        base_r, wt_r,
    )

    print("\n\u2713 All done. Outputs saved to ./outputs/")


if __name__ == "__main__":
    main()
```

## Data Augmentation and Hard Negative Mining

- **NEI is kept as a separate third class** (`supports`, `refutes`, `neutral`) instead of forcing the task into binary classification.
- **A warm-up model is trained first** on the original data so the pipeline has a baseline model for identifying difficult NEI cases.
- **Hard NEI mining is used**: the code looks only at true NEI examples and ranks them by how confusing they are for the model.
- **The hardest NEI examples are kept**: examples with high lexical overlap, shared entities, and low NEI confidence are prioritized.
- **Synthetic NEI augmentation is used lightly**: only a small fraction of NEI rows are sampled, and they are paired with evidence from other examples to create more NEI-like training cases.
- **The synthetic NEI generator avoids totally random pairs**: it keeps examples that are somewhat related but not decisive, so the model learns that *related does not mean supports*.
- **Class weights are boosted for NEI** so the model gives the neutral class more importance during training.
- **Training is selected using a NEI-focused score** rather than plain accuracy or loss alone. This score rewards neutral F1, neutral recall, macro F1, and accuracy.
- **Early stopping is based on the NEI-focused score**, so training stops when NEI behavior stops improving.
- **Regularization is added through dropout, label smoothing, and weight decay** to reduce overfitting while learning NEI.
- **Temperature scaling is applied after training** to improve confidence calibration, which helps make NEI decisions less overconfident.
- **The final script compares warm-up vs. NEI-tuned performance** so you can see whether NEI handling improved, not just overall accuracy.

```python
import os
import json
import random
import re
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

from datasets import load_dataset

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from torch.optim import AdamW

from transformers import (
    BertTokenizer,
    BertConfig,
    BertForSequenceClassification,
    get_linear_schedule_with_warmup,
)

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score,
    recall_score,
)
from sklearn.calibration import calibration_curve
from scipy.stats import pearsonr

warnings.filterwarnings("ignore")

# ==============================================================================
# Config
# ==============================================================================
SEED = 42
MAX_LEN = 128
BATCH_SIZE = 16

WARMUP_EPOCHS = 2
DEFAULT_FINAL_EPOCHS = 4

BASE_LR = 1e-5
BASE_WEIGHT_DECAY = 0.05
BASE_DROPOUT = 0.30
BASE_LABEL_SMOOTHING = 0.10

EARLY_STOP_PATIENCE = 2
EARLY_STOP_MIN_DELTA = 1e-4

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

LABEL_NAMES = ["supports", "refutes", "neutral"]
ID2LABEL = {0: "supports", 1: "refutes", 2: "neutral"}

OUTPUT_DIR = "./outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed(SEED)

# ==============================================================================
# Data Loading / Cleaning
# ==============================================================================
def normalize_pubhealth_label(label):
    label = str(label).strip().lower()
    if label == "true":
        return 0
    if label == "false":
        return 1
    if label in {"mixture", "unproven"}:
        return 2
    return None

def convert_pubhealth_split(df):
    out = df.copy()
    out["claim"] = out["text_1"].astype(str).str.strip()
    out["evidence"] = out["text_2"].astype(str).str.strip()
    out["label"] = out["label"].apply(normalize_pubhealth_label)

    out = out.dropna(subset=["claim", "evidence", "label"]).copy()
    out["label"] = out["label"].astype(int)

    return out[["claim", "evidence", "label"]].reset_index(drop=True)

def compute_class_weights(df, nei_boost=1.0):
    counts = Counter(df["label"])
    weights = []
    for i in range(len(LABEL_NAMES)):
        n = counts.get(i, 0)
        if n > 0:
            w = len(df) / (len(LABEL_NAMES) * n)
        else:
            w = 0.0
        if i == 2:
            w *= nei_boost
        weights.append(w)
    return torch.tensor(weights, dtype=torch.float)

# ==============================================================================
# Dataset
# ==============================================================================
class PubHealthDataset(Dataset):
    """
    Input to BERT: [CLS] claim [SEP] evidence [SEP]
    """
    def __init__(self, df: pd.DataFrame, tokenizer, max_len=MAX_LEN):
        self.df = df.reset_index(drop=True)
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        claim = str(row["claim"])
        evidence = str(row["evidence"])
        label = int(row["label"])

        enc = self.tokenizer(
            claim,
            evidence,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_tensors="pt",
        )

        return {
            "input_ids": enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "token_type_ids": enc["token_type_ids"].squeeze(0),
            "label": torch.tensor(label, dtype=torch.long),
            "claim": claim,
            "evidence": evidence,
            "idx": idx,
        }

# ==============================================================================
# Shortcut Features
# ==============================================================================
def lexical_overlap(claim: str, evidence: str) -> float:
    c_words = set(claim.lower().split())
    e_words = set(evidence.lower().split())
    if not e_words:
        return 0.0
    return len(c_words & e_words) / len(e_words)

def simple_entity_set(text: str):
    tokens = re.findall(r"\b[A-Z][a-zA-Z0-9\-]*\b", text)
    return set(tokens)

def entity_overlap(claim: str, evidence: str) -> float:
    c = simple_entity_set(claim)
    e = simple_entity_set(evidence)
    if not (c or e):
        return 0.0
    return len(c & e) / len(c | e)

def has_negation(text: str) -> int:
    neg_words = {
        "not", "no", "never", "neither", "nor", "n't",
        "nobody", "nothing", "none", "without", "lack", "lacking"
    }
    words = set(text.lower().split())
    return int(bool(words & neg_words))

def compute_shortcut_features(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out["overlap"] = out.apply(lambda r: lexical_overlap(r["claim"], r["evidence"]), axis=1)
    out["ent_overlap"] = out.apply(lambda r: entity_overlap(r["claim"], r["evidence"]), axis=1)
    out["negation"] = out["evidence"].apply(has_negation)
    out["claim_len"] = out["claim"].apply(lambda x: len(str(x).split()))
    out["evid_len"] = out["evidence"].apply(lambda x: len(str(x).split()))
    out["label_name"] = out["label"].map(ID2LABEL)
    return out

# ==============================================================================
# Scoring for NEI-focused selection
# ==============================================================================
def nei_priority_score(y_true, y_pred):
    per_class_f1 = f1_score(y_true, y_pred, average=None, labels=[0, 1, 2], zero_division=0)
    per_class_recall = recall_score(y_true, y_pred, average=None, labels=[0, 1, 2], zero_division=0)
    macro_f1 = f1_score(y_true, y_pred, average="macro", labels=[0, 1, 2], zero_division=0)
    acc = accuracy_score(y_true, y_pred)

    neutral_f1 = per_class_f1[2]
    neutral_recall = per_class_recall[2]
    return 0.45 * neutral_f1 + 0.25 * neutral_recall + 0.20 * macro_f1 + 0.10 * acc

# ==============================================================================
# Training / Evaluation
# ==============================================================================
def train_epoch(model, loader, optimizer, scheduler, criterion, device):
    model.train()
    total_loss = 0.0
    correct = 0
    total = 0

    for batch in loader:
        ids = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        ttype = batch["token_type_ids"].to(device)
        labs = batch["label"].to(device)

        optimizer.zero_grad()
        logits = model(
            input_ids=ids,
            attention_mask=mask,
            token_type_ids=ttype
        ).logits

        loss = criterion(logits, labs)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()

        total_loss += loss.item()
        correct += (logits.argmax(-1) == labs).sum().item()
        total += labs.size(0)

    return total_loss / max(1, len(loader)), correct / max(1, total)

def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss = 0.0
    correct = 0
    total = 0
    all_preds, all_labels, all_probs = [], [], []

    with torch.no_grad():
        for batch in loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)
            labs = batch["label"].to(device)

            logits = model(
                input_ids=ids,
                attention_mask=mask,
                token_type_ids=ttype
            ).logits

            probs = torch.softmax(logits, dim=-1)
            loss = criterion(logits, labs)

            total_loss += loss.item()
            preds = logits.argmax(-1)
            correct += (preds == labs).sum().item()
            total += labs.size(0)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labs.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())

    return (
        total_loss / max(1, len(loader)),
        correct / max(1, total),
        np.array(all_preds),
        np.array(all_labels),
        np.array(all_probs),
    )

def run_training(
    train_ds,
    val_ds,
    class_weights=None,
    tag="model",
    epochs=4,
    lr=BASE_LR,
    weight_decay=BASE_WEIGHT_DECAY,
    dropout=BASE_DROPOUT,
    label_smoothing=BASE_LABEL_SMOOTHING,
    patience=EARLY_STOP_PATIENCE,
    min_delta=EARLY_STOP_MIN_DELTA,
):
    """
    Fine-tune BERT with:
    - configurable dropout
    - optional class weights
    - label smoothing
    - early stopping on NEI-aware score
    - restore best checkpoint
    """
    print(f"\n{'='*60}")
    print(f"Training: {tag}")
    print(f"{'='*60}")

    config = BertConfig.from_pretrained(
        "bert-base-uncased",
        num_labels=3,
        hidden_dropout_prob=dropout,
        attention_probs_dropout_prob=dropout,
    )

    model = BertForSequenceClassification.from_pretrained(
        "bert-base-uncased",
        config=config
    ).to(DEVICE)

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    if class_weights is not None:
        criterion = nn.CrossEntropyLoss(
            weight=class_weights.to(DEVICE),
            label_smoothing=label_smoothing,
        )
        print(f"Class weights: {[f'{w:.3f}' for w in class_weights.tolist()]}")
    else:
        criterion = nn.CrossEntropyLoss(label_smoothing=label_smoothing)

    optimizer = AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    total_steps = len(train_loader) * epochs
    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=max(1, int(0.1 * total_steps)),
        num_training_steps=max(1, total_steps),
    )

    history = []
    best_state = None
    best_epoch = -1
    best_score = -1.0
    best_vl_loss = float("inf")
    best_preds = None
    best_labels = None
    best_probs = None
    bad_epochs = 0

    for epoch in range(1, epochs + 1):
        tr_loss, tr_acc = train_epoch(model, train_loader, optimizer, scheduler, criterion, DEVICE)
        vl_loss, vl_acc, vl_preds, vl_labels, vl_probs = evaluate(model, val_loader, criterion, DEVICE)

        macro_f1 = f1_score(vl_labels, vl_preds, average="macro", labels=[0, 1, 2], zero_division=0)
        per_class_f1 = f1_score(vl_labels, vl_preds, average=None, labels=[0, 1, 2], zero_division=0)
        neutral_f1 = per_class_f1[2]
        neutral_recall = recall_score(vl_labels, vl_preds, average=None, labels=[0, 1, 2], zero_division=0)[2]
        score = nei_priority_score(vl_labels, vl_preds)

        history.append({
            "epoch": epoch,
            "tr_loss": tr_loss,
            "tr_acc": tr_acc,
            "vl_loss": vl_loss,
            "vl_acc": vl_acc,
            "vl_macro_f1": macro_f1,
            "vl_neutral_f1": neutral_f1,
            "vl_neutral_recall": neutral_recall,
            "nei_score": score,
        })

        print(
            f"Epoch {epoch}/{epochs} | "
            f"Train Loss {tr_loss:.4f} Acc {tr_acc:.4f} | "
            f"Val Loss {vl_loss:.4f} Acc {vl_acc:.4f} | "
            f"Macro F1 {macro_f1:.4f} | "
            f"NEI F1 {neutral_f1:.4f} | "
            f"NEI Recall {neutral_recall:.4f} | "
            f"NEI Score {score:.4f}"
        )

        improved = (score > best_score + min_delta) or (
            abs(score - best_score) <= min_delta and vl_loss < best_vl_loss - min_delta
        )

        if improved:
            best_score = score
            best_vl_loss = vl_loss
            best_epoch = epoch
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            best_preds = vl_preds.copy()
            best_labels = vl_labels.copy()
            best_probs = vl_probs.copy()
            bad_epochs = 0
        else:
            bad_epochs += 1
            if bad_epochs >= patience:
                print(f"Early stopping triggered at epoch {epoch}. Best epoch was {best_epoch}.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    save_path = os.path.join(OUTPUT_DIR, f"bert_{tag}")
    model.save_pretrained(save_path)
    print(f"Model saved -> {save_path}")
    print(f"Best validation epoch -> {best_epoch} | Best NEI score -> {best_score:.4f} | Best val loss -> {best_vl_loss:.4f}")

    return model, best_preds, best_labels, best_probs, history, best_epoch, save_path

# ==============================================================================
# Temperature Calibration
# ==============================================================================
class TemperatureScaler(nn.Module):
    def __init__(self):
        super().__init__()
        self.log_temperature = nn.Parameter(torch.zeros(1))

    @property
    def temperature(self):
        return torch.exp(self.log_temperature)

    def forward(self, logits):
        return logits / self.temperature

    def fit(self, model, val_loader, device, lr=0.01, max_iter=50):
        self.cpu()
        model.eval()

        logits_list, labels_list = [], []
        with torch.no_grad():
            for batch in val_loader:
                ids = batch["input_ids"].to(device)
                mask = batch["attention_mask"].to(device)
                ttype = batch["token_type_ids"].to(device)
                labs = batch["label"].to(device)

                out = model(
                    input_ids=ids,
                    attention_mask=mask,
                    token_type_ids=ttype
                ).logits

                logits_list.append(out.cpu())
                labels_list.append(labs.cpu())

        logits_all = torch.cat(logits_list)
        labels_all = torch.cat(labels_list)

        criterion = nn.CrossEntropyLoss()
        opt = torch.optim.LBFGS([self.log_temperature], lr=lr, max_iter=max_iter)

        def step():
            opt.zero_grad()
            loss = criterion(self.forward(logits_all), labels_all)
            loss.backward()
            return loss

        opt.step(step)

        T = self.temperature.item()
        print(f"Learned temperature T = {T:.4f} (T > 1 usually means overconfidence)")

        cal_probs = torch.softmax(self.forward(logits_all).detach(), dim=-1).numpy()
        cal_preds = cal_probs.argmax(axis=-1)
        return cal_preds, labels_all.numpy(), cal_probs, T

# ==============================================================================
# NEI Hard Mining + Augmentation
# ==============================================================================
def hard_nei_score(overlap, ent_overlap, p_support, p_refute, p_nei):
    confusion = max(p_support, p_refute) - p_nei
    return 0.6 * confusion + 0.25 * overlap + 0.15 * ent_overlap

def mine_hard_nei_examples(model, df, tokenizer, device, top_frac=0.25, max_len=MAX_LEN):
    ds = PubHealthDataset(df, tokenizer, max_len=max_len)
    loader = DataLoader(ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    model.eval()
    rows = []

    with torch.no_grad():
        for batch in loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)
            labs = batch["label"].cpu().numpy()
            claims = batch["claim"]
            evids = batch["evidence"]

            logits = model(
                input_ids=ids,
                attention_mask=mask,
                token_type_ids=ttype
            ).logits

            probs = torch.softmax(logits, dim=-1).cpu().numpy()

            for i in range(len(labs)):
                if labs[i] != 2:
                    continue

                p_sup = float(probs[i, 0])
                p_ref = float(probs[i, 1])
                p_nei = float(probs[i, 2])

                ov = lexical_overlap(claims[i], evids[i])
                ent_ov = entity_overlap(claims[i], evids[i])
                score = hard_nei_score(ov, ent_ov, p_sup, p_ref, p_nei)

                rows.append({
                    "claim": claims[i],
                    "evidence": evids[i],
                    "label": 2,
                    "overlap": ov,
                    "entity_overlap": ent_ov,
                    "p_support": p_sup,
                    "p_refute": p_ref,
                    "p_nei": p_nei,
                    "hard_score": score,
                })

    hard_df = pd.DataFrame(rows)
    if len(hard_df) == 0:
        return hard_df

    hard_df = hard_df.sort_values("hard_score", ascending=False).reset_index(drop=True)
    keep_n = max(1, int(len(hard_df) * top_frac))
    return hard_df.head(keep_n)[["claim", "evidence", "label"]].reset_index(drop=True)

def augment_nei_examples(df, sample_frac=0.25, n_per_selected=1, max_relatedness=0.35, seed=SEED):
    """
    Synthetic NEI augmentation:
    - only sample a fraction of NEI rows
    - pair with evidence from another example
    - keep examples that are moderately related, but not too similar
    """
    rng = random.Random(seed)
    evidence_pool = df["evidence"].tolist()

    nei_rows = df[df["label"] == 2].reset_index(drop=True)
    if len(nei_rows) == 0:
        return pd.DataFrame(columns=["claim", "evidence", "label"])

    sample_size = max(1, int(len(nei_rows) * sample_frac))
    selected_idx = rng.sample(list(range(len(nei_rows))), k=sample_size)
    selected = nei_rows.iloc[selected_idx].reset_index(drop=True)

    aug_rows = []
    for _, row in selected.iterrows():
        claim = row["claim"]

        for _ in range(n_per_selected):
            best_evid = None
            best_score = -1.0

            for _trial in range(12):
                evid = rng.choice(evidence_pool)
                ov = lexical_overlap(claim, evid)
                ent_ov = entity_overlap(claim, evid)
                relatedness = 0.7 * ov + 0.3 * ent_ov

                if relatedness > best_score and relatedness < max_relatedness:
                    best_score = relatedness
                    best_evid = evid

            if best_evid is None:
                best_evid = rng.choice(evidence_pool)

            aug_rows.append({
                "claim": claim,
                "evidence": best_evid,
                "label": 2
            })

    return pd.DataFrame(aug_rows)

def build_augmented_train_df(train_df, hard_nei_df=None, synthetic_nei_df=None):
    parts = [train_df]
    if hard_nei_df is not None and len(hard_nei_df) > 0:
        parts.append(hard_nei_df)
    if synthetic_nei_df is not None and len(synthetic_nei_df) > 0:
        parts.append(synthetic_nei_df)

    aug_df = pd.concat(parts, ignore_index=True)
    aug_df = aug_df.sample(frac=1.0, random_state=SEED).reset_index(drop=True)
    return aug_df

# ==============================================================================
# NEI-focused Candidate Search
# ==============================================================================
NEI_TUNE_CONFIGS = [
    {
        "name": "nei_a",
        "lr": 1e-5,
        "weight_decay": 0.05,
        "dropout": 0.30,
        "label_smoothing": 0.10,
        "nei_boost": 1.50,
        "hard_nei_top_frac": 0.25,
        "synthetic_nei_sample_frac": 0.15,
        "synthetic_max_relatedness": 0.35,
        "epochs": 4,
    },
    {
        "name": "nei_b",
        "lr": 8e-6,
        "weight_decay": 0.08,
        "dropout": 0.20,
        "label_smoothing": 0.05,
        "nei_boost": 1.75,
        "hard_nei_top_frac": 0.30,
        "synthetic_nei_sample_frac": 0.10,
        "synthetic_max_relatedness": 0.30,
        "epochs": 5,
    },
    {
        "name": "nei_c",
        "lr": 5e-6,
        "weight_decay": 0.10,
        "dropout": 0.25,
        "label_smoothing": 0.05,
        "nei_boost": 2.00,
        "hard_nei_top_frac": 0.35,
        "synthetic_nei_sample_frac": 0.10,
        "synthetic_max_relatedness": 0.25,
        "epochs": 5,
    },
]

def train_nei_candidate(cfg, warm_model, train_df, dev_ds, tokenizer):
    print(f"\n--- Running config: {cfg['name']} ---")

    hard_nei_df = mine_hard_nei_examples(
        warm_model,
        train_df,
        tokenizer,
        DEVICE,
        top_frac=cfg["hard_nei_top_frac"],
    )

    synthetic_nei_df = augment_nei_examples(
        train_df,
        sample_frac=cfg["synthetic_nei_sample_frac"],
        n_per_selected=1,
        max_relatedness=cfg["synthetic_max_relatedness"],
    )

    aug_train_df = build_augmented_train_df(train_df, hard_nei_df, synthetic_nei_df)
    aug_weights = compute_class_weights(aug_train_df, nei_boost=cfg["nei_boost"])
    aug_train_ds = PubHealthDataset(aug_train_df, tokenizer)

    model, preds, labels, probs, history, best_epoch, save_path = run_training(
        train_ds=aug_train_ds,
        val_ds=dev_ds,
        class_weights=aug_weights,
        tag=cfg["name"],
        epochs=cfg["epochs"],
        lr=cfg["lr"],
        weight_decay=cfg["weight_decay"],
        dropout=cfg["dropout"],
        label_smoothing=cfg["label_smoothing"],
        patience=EARLY_STOP_PATIENCE,
    )

    per_class_f1 = f1_score(labels, preds, average=None, labels=[0, 1, 2], zero_division=0)
    neutral_f1 = float(per_class_f1[2])
    neutral_recall = float(recall_score(labels, preds, average=None, labels=[0, 1, 2], zero_division=0)[2])
    macro_f1 = float(f1_score(labels, preds, average="macro", labels=[0, 1, 2], zero_division=0))
    acc = float(accuracy_score(labels, preds))
    score = float(nei_priority_score(labels, preds))

    summary = {
        "cfg": cfg,
        "save_path": save_path,
        "best_epoch": best_epoch,
        "dev_accuracy": acc,
        "dev_macro_f1": macro_f1,
        "dev_f1_per_class": per_class_f1.tolist(),
        "neutral_f1": neutral_f1,
        "neutral_recall": neutral_recall,
        "nei_score": score,
        "hard_nei_kept": len(hard_nei_df),
        "synthetic_nei_created": len(synthetic_nei_df),
        "aug_train_n": len(aug_train_df),
        "aug_train_df": aug_train_df,
        "hard_nei_df": hard_nei_df,
        "synthetic_nei_df": synthetic_nei_df,
    }

    del model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return summary

# ==============================================================================
# Shortcut Analysis
# ==============================================================================
def add_overlap_bucket(df):
    out = df.copy()
    if out["overlap"].nunique() >= 4:
        out["overlap_bucket"] = pd.qcut(
            out["overlap"],
            q=4,
            duplicates="drop"
        )
    else:
        out["overlap_bucket"] = "All"
    return out

def shortcut_analysis(model, loader, val_df, device, tag="model"):
    model.eval()
    rows = []

    with torch.no_grad():
        for batch in loader:
            ids = batch["input_ids"].to(device)
            mask = batch["attention_mask"].to(device)
            ttype = batch["token_type_ids"].to(device)
            labs = batch["label"].to(device)
            claims = batch["claim"]
            evids = batch["evidence"]

            logits = model(
                input_ids=ids,
                attention_mask=mask,
                token_type_ids=ttype
            ).logits

            probs = torch.softmax(logits, dim=-1).cpu().numpy()
            preds = probs.argmax(axis=-1)
            labs_np = labs.cpu().numpy()

            for i in range(len(labs_np)):
                rows.append({
                    "label": labs_np[i],
                    "pred": preds[i],
                    "confidence": float(probs[i].max()),
                    "overlap": lexical_overlap(claims[i], evids[i]),
                    "entity_overlap": entity_overlap(claims[i], evids[i]),
                    "correct": int(preds[i] == labs_np[i]),
                    "tag": tag,
                })

    df = pd.DataFrame(rows)
    df["label_name"] = df["label"].map(ID2LABEL)
    df = add_overlap_bucket(df)

    if df["overlap"].nunique() > 1 and df["confidence"].nunique() > 1:
        r, p = pearsonr(df["overlap"], df["confidence"])
    else:
        r, p = np.nan, np.nan

    print(f"[{tag}] Overlap-Confidence Pearson r = {r} (p = {p})")
    if pd.notna(r) and r > 0.1:
        print("  Warning: confidence rises with lexical overlap -> possible shortcut reliance")
    else:
        print("  Weak overlap-confidence link")

    print(f"\n[{tag}] Accuracy by overlap bucket:")
    print(df.groupby("overlap_bucket")["correct"].mean().to_string())

    print(f"\n[{tag}] Accuracy by class:")
    print(df.groupby("label_name")["correct"].mean().to_string())

    return df, r

# ==============================================================================
# Plotting
# ==============================================================================
def plot_all(
    raw_df,
    warm_df, warm_preds, warm_labels, warm_probs,
    final_df, final_preds, final_labels, final_probs,
    cal_preds, cal_labels, cal_probs,
    warm_r, final_r,
):
    COLORS = ["steelblue", "tomato", "orange"]

    fig, axes = plt.subplots(3, 3, figsize=(18, 15))
    fig.suptitle(
        "PubHealth NEI Handling: Warm-up vs Best NEI-Tuned Model",
        fontsize=13,
        fontweight="bold",
    )

    ax = axes[0, 0]
    counts = [Counter(raw_df["label"])[i] for i in range(3)]
    bars = ax.bar(LABEL_NAMES, counts, color=COLORS)
    for b, c in zip(bars, counts):
        ax.text(b.get_x() + b.get_width() / 2, b.get_height() + 10, str(c), ha="center", fontsize=9)
    ax.set_title("Label Distribution")
    ax.set_ylabel("Count")
    ax.tick_params(axis="x", rotation=15)

    ax = axes[0, 1]
    for name, color in zip(LABEL_NAMES, COLORS):
        sub = raw_df[raw_df["label_name"] == name]["overlap"]
        ax.hist(sub, bins=30, alpha=0.6, label=name, color=color, density=True)
    ax.set_title("Lexical Overlap by Label")
    ax.set_xlabel("Overlap Score")
    ax.set_ylabel("Density")
    ax.legend(fontsize=8)

    ax = axes[0, 2]
    for name, color in zip(LABEL_NAMES, COLORS):
        sub = warm_df[warm_df["label_name"] == name]
        ax.scatter(sub["overlap"], sub["confidence"], alpha=0.25, s=8, label=name, color=color)
    ax.set_title(f"Warm-up: Confidence vs Overlap\n(r = {warm_r:.3f})")
    ax.set_xlabel("Lexical Overlap")
    ax.set_ylabel("Model Confidence")
    ax.legend(markerscale=3, fontsize=8)

    ax = axes[1, 0]
    warm_tmp = warm_df.copy()
    final_tmp = final_df.copy()
    warm_tmp["bucket_str"] = warm_tmp["overlap_bucket"].astype(str)
    final_tmp["bucket_str"] = final_tmp["overlap_bucket"].astype(str)

    bucket_names = list(pd.Index(warm_tmp["bucket_str"].tolist() + final_tmp["bucket_str"].tolist()).unique())

    warm_b = warm_tmp.groupby("bucket_str")["correct"].mean().reindex(bucket_names)
    final_b = final_tmp.groupby("bucket_str")["correct"].mean().reindex(bucket_names)

    x = np.arange(len(bucket_names))
    ax.bar(x - 0.2, warm_b.values, 0.4, label="Warm-up", color="steelblue")
    ax.bar(x + 0.2, final_b.values, 0.4, label="Final", color="coral")
    ax.set_xticks(x)
    ax.set_xticklabels(bucket_names, rotation=15)
    ax.set_title("Accuracy by Overlap Bucket")
    ax.set_ylabel("Accuracy")
    ax.legend()
    ax.set_ylim(0, 1)

    ax = axes[1, 1]
    cm = confusion_matrix(warm_labels, warm_preds)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=ax,
                xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES)
    ax.set_title("Confusion Matrix: Warm-up")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.tick_params(axis="x", rotation=20)

    ax = axes[1, 2]
    cm_w = confusion_matrix(final_labels, final_preds)
    sns.heatmap(cm_w, annot=True, fmt="d", cmap="Oranges", ax=ax,
                xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES)
    ax.set_title("Confusion Matrix: Final NEI-Tuned")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")
    ax.tick_params(axis="x", rotation=20)

    ax = axes[2, 0]
    warm_f1 = f1_score(warm_labels, warm_preds, average=None, labels=[0, 1, 2], zero_division=0)
    final_f1 = f1_score(final_labels, final_preds, average=None, labels=[0, 1, 2], zero_division=0)
    cal_f1 = f1_score(cal_labels, cal_preds, average=None, labels=[0, 1, 2], zero_division=0)

    x = np.arange(3)
    ax.bar(x - 0.27, warm_f1, 0.27, label="Warm-up", color="steelblue")
    ax.bar(x, final_f1, 0.27, label="Final", color="coral")
    ax.bar(x + 0.27, cal_f1, 0.27, label="Calibrated", color="mediumseagreen")
    ax.set_xticks(x)
    ax.set_xticklabels(LABEL_NAMES, rotation=10)
    ax.set_title("Per-Class F1")
    ax.set_ylabel("F1 Score")
    ax.legend(fontsize=8)
    ax.set_ylim(0, 1)

    ax = axes[2, 1]
    for cls_id, (name, color) in enumerate(zip(LABEL_NAMES, COLORS)):
        bl = (warm_labels == cls_id).astype(int)
        try:
            fp, mp = calibration_curve(bl, warm_probs[:, cls_id], n_bins=10)
            ax.plot(mp, fp, marker="o", label=name, color=color)
        except ValueError:
            pass
    ax.plot([0, 1], [0, 1], "k--", label="Perfect")
    ax.set_title("Calibration Curve: Warm-up")
    ax.set_xlabel("Mean Predicted Probability")
    ax.set_ylabel("Fraction of Positives")
    ax.legend(fontsize=8)

    ax = axes[2, 2]
    for cls_id, (name, color) in enumerate(zip(LABEL_NAMES, COLORS)):
        bl = (cal_labels == cls_id).astype(int)
        try:
            fp, mp = calibration_curve(bl, cal_probs[:, cls_id], n_bins=10)
            ax.plot(mp, fp, marker="o", label=name, color=color)
        except ValueError:
            pass
    ax.plot([0, 1], [0, 1], "k--", label="Perfect")
    ax.set_title("Calibration: After Temperature Scaling")
    ax.set_xlabel("Mean Predicted Probability")
    ax.set_ylabel("Fraction of Positives")
    ax.legend(fontsize=8)

    plt.tight_layout()
    out = os.path.join(OUTPUT_DIR, "nei_augmented_pubhealth_best.png")
    plt.savefig(out, dpi=150, bbox_inches="tight")
    print(f"\nFigure saved -> {out}")
    plt.show()

# ==============================================================================
# Main
# ==============================================================================
def main():
    print(f"Device: {DEVICE}\n")

    dataset = load_dataset("bigbio/pubhealth", "pubhealth_bigbio_pairs")

    train_df = convert_pubhealth_split(dataset["train"].to_pandas())
    dev_df = convert_pubhealth_split(dataset["validation"].to_pandas())
    test_df = convert_pubhealth_split(dataset["test"].to_pandas())

    for name, df_ in [("Train", train_df), ("Dev", dev_df), ("Test", test_df)]:
        print(f"\n{name} label distribution:")
        print(df_["label"].value_counts().sort_index())

    print("\nPubHealth loaded:")
    print(f"  Train: {len(train_df)}")
    print(f"  Dev:   {len(dev_df)}")
    print(f"  Test:  {len(test_df)}")

    tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

    train_ds = PubHealthDataset(train_df, tokenizer)
    dev_ds = PubHealthDataset(dev_df, tokenizer)
    test_ds = PubHealthDataset(test_df, tokenizer)

    dev_loader = DataLoader(dev_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    # --------------------------------------------------------------------------
    # Raw shortcut features on dev set
    # --------------------------------------------------------------------------
    print("\nComputing raw shortcut features on dev set...")
    raw_df = compute_shortcut_features(dev_df)

    print("\nMean lexical overlap by label:")
    print(raw_df.groupby("label_name")["overlap"].mean().to_string())

    print("\nMean entity overlap by label:")
    print(raw_df.groupby("label_name")["ent_overlap"].mean().to_string())

    nei_raw = raw_df[raw_df["label"] == 2]
    print("\nNEI-only stats on dev:")
    print(f"  Count: {len(nei_raw)}")
    if len(nei_raw) > 0:
        print(f"  Mean overlap: {nei_raw['overlap'].mean():.4f}")
        print(f"  Mean entity overlap: {nei_raw['ent_overlap'].mean():.4f}")
        print(f"  Mean evidence length: {nei_raw['evid_len'].mean():.4f}")

    print("\nNegation rate by label:")
    print(raw_df.groupby("label_name")["negation"].mean().to_string())

    print("\nMean evidence length by label (tokens):")
    print(raw_df.groupby("label_name")["evid_len"].mean().to_string())

    # --------------------------------------------------------------------------
    # 1) Warm-up model on original train set
    # --------------------------------------------------------------------------
    warm_model, warm_preds, warm_labels, warm_probs, _, warm_best_epoch, warm_save_path = run_training(
        train_ds=train_ds,
        val_ds=dev_ds,
        class_weights=None,
        tag="warmup",
        epochs=WARMUP_EPOCHS,
        lr=BASE_LR,
        weight_decay=BASE_WEIGHT_DECAY,
        dropout=BASE_DROPOUT,
        label_smoothing=BASE_LABEL_SMOOTHING,
        patience=EARLY_STOP_PATIENCE,
    )

    # --------------------------------------------------------------------------
    # 2) Search NEI-focused configs
    # --------------------------------------------------------------------------
    print("\nSearching NEI-focused hyperparameter settings...")
    candidate_summaries = []
    best_candidate = None

    for cfg in NEI_TUNE_CONFIGS:
        cand = train_nei_candidate(cfg, warm_model, train_df, dev_ds, tokenizer)
        candidate_summaries.append({
            "name": cfg["name"],
            "lr": cfg["lr"],
            "weight_decay": cfg["weight_decay"],
            "dropout": cfg["dropout"],
            "label_smoothing": cfg["label_smoothing"],
            "nei_boost": cfg["nei_boost"],
            "hard_nei_top_frac": cfg["hard_nei_top_frac"],
            "synthetic_nei_sample_frac": cfg["synthetic_nei_sample_frac"],
            "synthetic_max_relatedness": cfg["synthetic_max_relatedness"],
            "epochs": cfg["epochs"],
            "best_epoch": cand["best_epoch"],
            "dev_accuracy": cand["dev_accuracy"],
            "dev_macro_f1": cand["dev_macro_f1"],
            "neutral_f1": cand["neutral_f1"],
            "neutral_recall": cand["neutral_recall"],
            "nei_score": cand["nei_score"],
            "hard_nei_kept": cand["hard_nei_kept"],
            "synthetic_nei_created": cand["synthetic_nei_created"],
            "aug_train_n": cand["aug_train_n"],
            "save_path": cand["save_path"],
        })

        print(
            f"Config {cfg['name']} -> "
            f"Acc={cand['dev_accuracy']:.4f}, "
            f"MacroF1={cand['dev_macro_f1']:.4f}, "
            f"NeutralF1={cand['neutral_f1']:.4f}, "
            f"NeutralRecall={cand['neutral_recall']:.4f}, "
            f"NEI Score={cand['nei_score']:.4f}"
        )

        if best_candidate is None or cand["nei_score"] > best_candidate["nei_score"]:
            best_candidate = cand

    print(f"\nBest config: {best_candidate['cfg']['name']}")
    print(f"Best NEI score: {best_candidate['nei_score']:.4f}")

    # Load best final model from disk
    final_model = BertForSequenceClassification.from_pretrained(best_candidate["save_path"]).to(DEVICE)

    # Recreate loaders / data from best candidate
    aug_train_df = best_candidate["aug_train_df"]
    hard_nei_df = best_candidate["hard_nei_df"]
    synthetic_nei_df = best_candidate["synthetic_nei_df"]

    # --------------------------------------------------------------------------
    # 3) Calibration on best final model
    # --------------------------------------------------------------------------
    print("\nFitting temperature scaling (post-hoc calibration)...")
    scaler = TemperatureScaler()
    cal_preds, cal_labels, cal_probs, T = scaler.fit(final_model, dev_loader, DEVICE)

    # --------------------------------------------------------------------------
    # 4) Shortcut analysis
    # --------------------------------------------------------------------------
    print("\nShortcut analysis - warm-up model:")
    warm_short_df, warm_r = shortcut_analysis(warm_model, dev_loader, dev_df, DEVICE, "warmup")

    print("\nShortcut analysis - best NEI-tuned model:")
    final_short_df, final_r = shortcut_analysis(final_model, dev_loader, dev_df, DEVICE, "best_nei_tuned")

    # --------------------------------------------------------------------------
    # 5) Dev + Test evaluation
    # --------------------------------------------------------------------------
    criterion = nn.CrossEntropyLoss(label_smoothing=BASE_LABEL_SMOOTHING)
    _, final_dev_acc, final_dev_preds, final_dev_labels, final_dev_probs = evaluate(
        final_model, dev_loader, criterion, DEVICE
    )

    print(f"\n{'─' * 55}")
    print("Warm-up")
    print(f"Overall Accuracy: {accuracy_score(warm_labels, warm_preds):.4f}")
    print(classification_report(warm_labels, warm_preds, labels=[0, 1, 2], target_names=LABEL_NAMES, zero_division=0))

    print(f"\n{'─' * 55}")
    print("Best NEI-Tuned")
    print(f"Overall Accuracy: {accuracy_score(final_dev_labels, final_dev_preds):.4f}")
    print(classification_report(final_dev_labels, final_dev_preds, labels=[0, 1, 2], target_names=LABEL_NAMES, zero_division=0))

    print(f"\n{'─' * 55}")
    print("Best NEI-Tuned + Calibration")
    print(f"Overall Accuracy: {accuracy_score(cal_labels, cal_preds):.4f}")
    print(classification_report(cal_labels, cal_preds, labels=[0, 1, 2], target_names=LABEL_NAMES, zero_division=0))

    print("\nFinal evaluation on test set (best NEI-tuned model):")
    test_loss, test_acc, test_preds, test_labels, test_probs = evaluate(
        final_model, test_loader, criterion, DEVICE
    )
    print(f"  Test Loss: {test_loss:.4f}")
    print(f"  Test Acc : {test_acc:.4f}")
    print(classification_report(test_labels, test_preds, labels=[0, 1, 2], target_names=LABEL_NAMES, zero_division=0))
    test_f1_per_class = f1_score(test_labels, test_preds, average=None, labels=[0, 1, 2], zero_division=0).tolist()

    # --------------------------------------------------------------------------
    # 6) Save summary
    # --------------------------------------------------------------------------
    results = {
        "dataset": "PubHealth",
        "model": "bert-base-uncased",
        "train_n_original": len(train_df),
        "train_n_augmented": len(aug_train_df),
        "dev_n": len(dev_df),
        "test_n": len(test_df),
        "best_config": {
            "name": best_candidate["cfg"]["name"],
            "lr": best_candidate["cfg"]["lr"],
            "weight_decay": best_candidate["cfg"]["weight_decay"],
            "dropout": best_candidate["cfg"]["dropout"],
            "label_smoothing": best_candidate["cfg"]["label_smoothing"],
            "nei_boost": best_candidate["cfg"]["nei_boost"],
            "hard_nei_top_frac": best_candidate["cfg"]["hard_nei_top_frac"],
            "synthetic_nei_sample_frac": best_candidate["cfg"]["synthetic_nei_sample_frac"],
            "synthetic_max_relatedness": best_candidate["cfg"]["synthetic_max_relatedness"],
            "epochs": best_candidate["cfg"]["epochs"],
        },
        "augmentation": {
            "hard_nei_kept": len(hard_nei_df),
            "synthetic_nei_created": len(synthetic_nei_df),
        },
        "candidate_summaries": candidate_summaries,
        "warmup": {
            "best_epoch": warm_best_epoch,
            "dev_accuracy": accuracy_score(warm_labels, warm_preds),
            "dev_f1_per_class": f1_score(warm_labels, warm_preds, average=None, labels=[0, 1, 2], zero_division=0).tolist(),
            "overlap_confidence_pearsonr": warm_r,
        },
        "final_nei_tuned": {
            "best_epoch": best_candidate["best_epoch"],
            "dev_accuracy": accuracy_score(final_dev_labels, final_dev_preds),
            "dev_f1_per_class": f1_score(final_dev_labels, final_dev_preds, average=None, labels=[0, 1, 2], zero_division=0).tolist(),
            "overlap_confidence_pearsonr": final_r,
            "test_accuracy": test_acc,
            "test_f1_per_class": test_f1_per_class,
            "nei_score": best_candidate["nei_score"],
            "neutral_f1": best_candidate["neutral_f1"],
            "neutral_recall": best_candidate["neutral_recall"],
        },
        "calibration": {
            "learned_temperature": T,
            "dev_accuracy": accuracy_score(cal_labels, cal_preds),
            "dev_f1_per_class": f1_score(cal_labels, cal_preds, average=None, labels=[0, 1, 2], zero_division=0).tolist(),
        },
    }

    json_path = os.path.join(OUTPUT_DIR, "results_summary_pubhealth_nei_tuned_final.json")
    with open(json_path, "w") as f:
        json.dump(results, f, indent=2)
    print(f"\nResults JSON -> {json_path}")

    # --------------------------------------------------------------------------
    # 7) Plots
    # --------------------------------------------------------------------------
    plot_all(
        raw_df,
        warm_short_df, warm_preds, warm_labels, warm_probs,
        final_short_df, final_dev_preds, final_dev_labels, final_dev_probs,
        cal_preds, cal_labels, cal_probs,
        warm_r, final_r,
    )

    print("\nDone. Outputs saved to ./outputs/")

if __name__ == "__main__":
    main()
```

  * F1: 0.39
  * Recall: 0.37
* **Observation:**

  * Model performs well on *supports* and *refutes*.
  * **Struggles to identify NEI**, with low recall and F1.
  * Indicates bias toward making definitive decisions rather than abstaining.

---

### **2. NEI-Tuned Model (Best Configuration: `nei_b`)**

* **Accuracy (Dev):** 64.60% (↓ from baseline)
* **Accuracy (Test):** 63.36%

#### **NEI Performance (Dev):**

* **F1:** 0.48 (↑ from 0.39)
* **Recall:** 0.88 (↑ significantly from 0.37)

#### **NEI Performance (Test):**

* **F1:** 0.51
* **Recall:** 0.87

#### **Other Classes:**

* **Supports:** remains strong (high precision, moderate recall)
* **Refutes:** significant drop in recall and F1

---

### **3. Key Trade-offs**

* **Major gain:**

  * The model learns to correctly identify NEI cases.
  * **NEI recall nearly doubles**, showing improved uncertainty detection.

* **Cost:**

  * Overall accuracy decreases (~4–5% drop).
  * Performance on *refutes* deteriorates significantly.
  * Model becomes more conservative, over-predicting NEI.

---

### **4. Effect of NEI-Specific Techniques**

* **Hard Negative Mining:**

  * Successfully exposed difficult NEI cases (high overlap, misleading evidence).
  * Improved boundary between NEI and other classes.

* **Synthetic NEI Augmentation:**

  * Helped model learn that *semantic similarity ≠ entailment*.
  * Contributed to higher NEI recall.

* **Class Weighting (NEI Boost):**

  * Strongly increased NEI sensitivity.
  * Led to **bias toward predicting NEI**, explaining drop in refutes performance.

* **NEI-Focused Model Selection:**

  * Using a custom NEI score prioritized neutral class performance.
  * Shifted optimization away from overall accuracy.

---

### **5. Shortcut Analysis Insights**

* Weak correlation between lexical overlap and model confidence:

  * Indicates **reduced reliance on shallow lexical shortcuts**.
* Accuracy still varies across overlap buckets:

  * Suggests partial robustness but not complete elimination of shortcuts.

---

### **6. Calibration**

* Temperature scaling yielded **T ≈ 1.10**:

  * Confirms mild overconfidence in predictions.
  * Calibration improved probability reliability but **did not change class predictions**.

---

### **7. Overall Conclusion**

* The proposed NEI-focused pipeline:

  * **Significantly improves NEI detection**, especially recall.
  * Encourages the model to **abstain when evidence is insufficient**.
* However:

  * This comes at the cost of **lower overall accuracy** and degraded performance on other classes.
  * The model becomes **over-sensitive to NEI**, suggesting overcompensation.

---

### **8. Key Takeaway**

> There is a fundamental trade-off between **accurate classification** and **reliable uncertainty detection (NEI)**.
> Improving NEI handling requires sacrificing some performance on definitive classes.
